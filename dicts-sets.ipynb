## 사전 

사전은 **키**<font size = "2">key</font>와 **값**<font size = "2">value</font>의
쌍으로 구성된 항목들로 구성된 모음 자료형이며, **딕셔너리**<font size = "2">dictionary</font> 라고도 불린다.

예를 들어,  `'Hello':'안녕'`은 `'Hello'`를 키로, `'안녕'`을 값으로 갖는 쌍이며,
`'world':'세계'`는 `'world'`를 키로, `'세계'`를 값으로 갖는 쌍이다.
두 쌍을 항목으로 갖는 사전은 다음과 같이 두 쌍을 중괄호로 감싼다.

```python
{'Hello':'안녕', 'world':'세계'}
```

**빈 사전**

빈 사전은 아무것도 포함하지 않는 사전을 의미한다. 
빈 사전를 만드는 방법은 아래와 같으며, 
중괄호를 사용한다.

````python
empty_dict = {}
empty_dict
````

다음 방식도 빈 사전을 생성한다.

````python
empty_dict = dict()
empty_dict
````

**`dict` 자료형**

사전은 `dict` 자료형의 값이다.

````python
type({'Hello':'안녕', 'world':'세계'})
````

**`dict()` 함수**

`dict()` 함수는 키와 값의 튜플 형식의 항목으로 구성된 모음 자료형을
사전 자료형으로 변환한다. 

예를 들어, 아래 코든느 튜플로 구성된 리스트를 사전으로 변환한다.
튜플의 첫째 항목은 키로, 둘째 항목은 값으로 사용된다.

````python
data = [('Hello', '안녕'), ('World', '세계'), ('Programming', '프로그래밍')]
dict(data)
````

`zip()` 함수와 `dict()` 함수를 조합하여 편리하게 사전을 생성할 수 있다.

````python
data = zip('abcde', [1, 2, 3, 4, 5])
data_dict = dict(data)
data_dict
````

### 사전의 키

사전의 키로 정수, 부동소수점, 문자열, 튜플 등 불변 자료형만 허용된다.
즉, 리스트와 같은 가변 자료형은 사전의 키로 사용될 수 없다.
반면에 키에 대응하는 값은 임의의 자료형이 허용된다.

- 정수, 부동소수점 키

````python
dic1 = {2: '정수', 1.2 : '부동소수점'}
dic1
````

- 문자열 키

````python
dic2 ={'a':1, 'b':2, 'c':3}
dic2
````

- 튜플 키

````python
dic3 = {(1, 'a') : [1, 2, 3], (2, 'b') : [4, 5, 6]}
dic3
````

- 리스트 키?

리스트를 키로 사용할 수는 없다는 의미에서 `TypeError`가 발생한다.

````python
dic4 = {[1, 2] : 'a'}
dic4
````

- 리스트를 포함하는 튜플은?

리스트와 같은 가변 자료형은 어떤 형식으로도 키에 포함되어서는 안된다.
아래 코드는 튜플을 키로 사용하지만 키의 첫째 항목이 리스트이기에 허용되지 않는다.

````python
dic4 = {([1, 2], '가') : 'ABC'}
dic4
````

### 사전 연산

문자열, 리스트, 튜플과는 다르게 사전 연산은 특별히 없다.
다만 항목 포함 여부와 항목의 개수를 확인해주는 `in` 연산자와 
`len()` 함수 정도만 활용된다.

**`in` 연산자**

사전의 키로 사용되었는 여부를 알려주는 논리 연산자다.
예를 들어, 아래 코드는 `'city'` 문자열이 사전의 키로 사용되었는지 여부를 판별한다.

````python
'city' in {"name": "강현", "age": "3"} 
````

`in`의 부정은 `not in`을 사용한다. 즉
키로 사용되지 않았는지 여부를 뭍는다.

````python
'city' not in {"name": "강현", "age": "3"} 
````

**`len()` 함수**

사전에 포함된 항목의 개수를 반환한다.

````python
len({"name": "강현", "age": "3"}) 
````

### 사전 인덱싱 

사전 인덱싱은 키를 활용한다.
사전은 비순차 자료형이기에 정수 인덱스를 지원하지 않으며 따라서 정수 인덱싱도 사용할 수 없다.

예를 들어, 아래 딕셔너리 `dic`에서 `'Hello'` 문자열을 키로 갖는 항목의 값을 확인하고자 하면 
다음과 같이 대괄호 안에 `'Hello'`를 지정한다.

````python
dic = {'Hello' : '안녕', 'World' : '세계'}
dic['Hello']
````

만약에 키로 사용되지 않는 값으로 인덱싱을 시도하면 
지정된 키가 없다는 `KeyError` 오류가 발생한다. 

````python
dic['Python']
````

사전은 가변 자료형이다. 
즉, 사전의 항목을 추가, 업데이트, 삭제할 수 있다.

**사전 항목 추가, 업데이트**

사전에 항목을 추가하거나 기존 항목을 업데이트는 사전 인덱싱 형식으로 이루어진다.

```python
사전[key] = value
```   

예를 들어, `dic`에 `Python : 파이썬` 항목을 다음과 같이 추가할 수 있다.

````python
dic['Python'] = '파이썬'
dic
````

만약에 키가 이미 사용되었다면 키와 연결된 값이 업데이트 된다.
예를 들어 아래 코드는 `'world'` 키의 값의 `'세계'`에서 `'세상'`으로 업데이트한다.

````python
dic['World'] = '세상'
dic
````

**사전 항목 삭제**

사전의 항목은 `del` 명령어에 키를 지정하여 삭제할 수 있다.
아래 코드는 `'world'` 가 키로 사용된 항목을 삭제한다.

````python
del dic['World']
dic
````

사용되지 않은 키를 지정해서 항목 삭제를 시도하면 
해당 키를 갖는 항목이 없다는 `KeyError` 오류가 발생한다.

````python
del dic['World']
dic
````

### 사전 반복문

사전에 대한 반복문은 키에 대해 실행된다.
아래 코드는 `data_dict` 사전에 사용된 키들만을 화면에 줄바꿈 없이 출력한다.

````python
data_dict = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}

for item in data_dict:
    print(item, end=' ')
````

**`items()` 메서드 활용**

키와 값의 쌍에 대해 반복문을 실행하려면 `items()` 메서드를 이용한다. 
단, 키와 값을 따로따로 화면에 출력하는 아래 코드에서처럼 튜플 해체 기법으로 키와 값 각각에 대해 변수를 지정하는 게 편리하다.

````python
for key, value in data_dict.items():
    print(f"{key:>8} 키의 값: {value}")
````

물론 튜플 해체를 반드시 사용할 필요는 없다.
다만 필요에 따라 튜플 인덱싱으로 키와 값을 구분에서 활용하면 된다.

````python
for item in data_dict.items():
    key = item[0]
    value = item[1]
    print(f"{key:>8} 키의 값: {value}")
````

**`values()` 메서드 활용**

아래 코드처럼 값만을 이용하려면 `values()` 메서드와 함께
반복문을 싱행한다.

````python
for item in data_dict.values():
    print(item, end=' ')
````

**예제 2**

정수들의 리스트가 다음과 같이 주어졌다.

````python
num_list = [2, 5, 2, 3, 3, 2]
````

(1) `num_list`에서 정수 2가 항목으로 위치한 인덱스들로 구성된 리스트를 가리키는 `idx_2` 변수를 정의하는 코드를 작성하라.

힌트: `enumerate()` 함수와 `for` 반복문 활용

답:

`enumerate()` 함수를 리스트에 적용하면 항목과 인덱스의 쌍으로 구성된 모음 자료형이 생성되고,
이 점을 활용하여 `for` 반복문을 실행하면 항목과 인덱스의 관계가 확인된다.

````python
for idx, item in enumerate(num_list):
    print(f"{idx} 번째 요소: {item}")
````

유사한 방식으로 정수 2가 위치한 곳의 인덱스만 추출하여 새로운 리스트를 생성하면 된다.

````python
idx_2 = []
for idx, item in enumerate(num_list):
    if item == 2:
        idx_2.append(idx)

print(idx_2)
````

(2) 위 리스트를 이용하여 아래와 같은 사전을 가리키는 `num_dict` 변수를 선언하는 코드를 작성하라.
사전에서 키는 리스트의 항목이 사용되며, 키의 값으로는 해당 항목이 위치한 곳의 인덱스로 구성된 리스트이다.

```python
{2: [0, 2, 5], 3: [3, 4], 5: [1]}
```

힌트: `enumerate()` 함수와 `for` 반복문

답:

따라서 

````python
def list2dic(xs):
    list_dict = dict()
    for i, item in enumerate(xs):
        if item in list_dict:
            list_dict[item].append(i)
        else:
            list_dict[item] = [i]

    return list_dict
````

````python
print(list2dic([2, 5, 2, 3, 3, 2]))
print(list2dic([15, 3, 15, 1, 3, 8]))
````

6명 각자의 이름을 키로, 나머지 정보로 구성된 리스트를 값으로 사용하는 사전을 구성한다.

먼저 `info_list`의 항목은 아래 형식임에 주의한다.

```python
person = ['김강현', '010-1234-5678', 20, 172.5, '제주']
```

위 리스트를 이름과 나머지로 구분하기 위해 다음과 같이 리스트 해체를 이용한다.

```python
name, *info = person
```

그러면 `name`은 이름을, `info`는 나머지 항목으로 구성된 리스트를 가리키게 되어

```python
name = person[0]

```

`info_dict[name] = info`

````python
info_dict = dict()

for person in info_list:
    name, *info = person
    info_dict[name] = info

info_dict
````

힌트: 사전 인덱싱, 리스트 인덱싱 활용

사전 인덱싱은 마치 하나의 함수처럼 작동한다.

김선주의 전화번호는 위 결과인 리스트에 대해 정수 인덱싱을 활용하면 바로 알 수 있다.

````python
def info_tel(name):
    return info_dict[name][0]
````

답:

이전 정의에서 `info_dict[name]` 표현식을

그리고 `if`조건문을 활용하여 `info_dict.get(name)`의 반환값이 `None`인 경우

````python
def info_tel(name):
    info = info_dict.get(name)
    if info:
        return info[0]
    else:
        return "정보 없음!"
````

**`hash()` 함수**

파이썬 객체의 해시 가능성은 `hash()` 함수의 인자로 사용될 수 있는가에 의해 결정된다.
즉 `hash()` 함수와 함께 호출됐을 때 오류가 발생하지 않아야 한다.
이렇게 `hash()` 함수와 함께 호출되었을 때 오류가 발생하지 않은 값을
**해시 가능**<font size="2">hashable</font>하다고 부른다.

해시 가능한 값에 대해 `hash()` 함수의 반환값은 특정 정수이며
서로 다른 두 값은 서로 다른 반환값을 갖는다.
즉, 조금이라도 다른 해시 가능한 값이 인자로 지정되면 다른 값을 계산한다.

````python
hash('123')
````

````python
hash('123 ')
````

````python
hash((1, 3, 4))
````

````python
hash((1, 2, 4))
````

:::{admonition} 해시 함수의 반환값
:class: note

`hash()` 함수의 반환값이 실행할 때마다 달라질 수 있다.
이는 보안상의 이유로 무작위 기능이 적용되기 때문이다.
항상 고정된 값을 반환하는 해시 함수도 있지만 여기서는 다루지 않는다.
:::

반면에 리스트 등 가변 자료형의 객체는 해시 가능하지 않다.

````python
hash([1, 3])
````

리스트를 포함한 튜플도 해시 가능하지 않다.

````python
hash(([1, 2], 3))
````

집합에 해시 가능한 값만 포함될 수 있는 이유는 두 항목을 구별하기 위해 `hash()` 함수를 이용하기 때문이다.
반면에 리스트 등과 같은 가변 자료형의 값은 언제든 변할 수 있기에 정체를
제대로 파악할 수 없다.

### 예제

**예제 7**

리스트를 인자로 받아서 사용된 항목의 개수를 반환하는 함수
`count_elem()`를 구현하라.
단, 중복 항목은 하나로 간주한다.

```
count_elem([2, 5, 2, 3, 3, 8, 2, 7]) = 5
count_elem([15, 3, 15, 1, 3]) = 3
```

답:

````python
def count_elem(xs):
    return len(set(xs))

print(count_elem([2, 5, 2, 3, 3, 8, 2, 7]))
print(count_elem([15, 3, 15, 1, 3]))
````

**문제 8**

리스트를 인자로 크기 내림차순으로 정렬한 리스트를 반환하는 함수
`sort_elem()`를 구현하라.
단, 중복 항목은 무시한다.

```
sort_elem([2, 5, 2, 3, 3, 8, 2, 7]) = [8, 7, 5, 3, 2]
sort_elem([15, 3, 15, 1, 3]) = [15, 3, 1]
```

답:

````python
def sort_elem(xs):
    return sorted(list(set(xs)), reverse=True)

print(sort_elem([2, 5, 2, 3, 3, 8, 2, 7]))
print(sort_elem([15, 3, 15, 1, 3]))
````

## 조건제시법

조건제시법<font size = "2">comprehension</font>을 이용하여 
리스트, 집합, 사전을 정의할 수 있다.

**리스트 조건제시법**

수학에서 0과 10사이에 있는 홀수들의 제곱을 원소로 갖는 집합을 조건제시법으로 표현하면
다음과 같다.

$$\{ x^2 \mid 0 \le x \le 10, \text{ 단 $x$는 홀수} \}$$

0과 10 사이에 있는 홀수들의 제곱을 항목으로 갖는 리스트를 `for` 반복문으로 구현해 보자.

````python
zero2ten_odd = []

for x in range(11):
    if x%2 == 1:
        zero2ten_odd.append(x**2)

zero2ten_odd
````

조건제시법을 이용하여 보다 간단하게 리스트를 생성할 수 있다.

````python
zero2ten_odd = [x**2 for x in range(11) if x%2 == 1]
zero2ten_odd
````

위 두 코드를 비교하면 조건제시법의 작동원리를 이해할 수 있을 것이다. 

**집합 조건제시법**

위 결과를 집합으로 구현하면 다음과 같다.

````python
zero2ten_odd_set = {x**2 for x in range(11) if x%2 == 1}
zero2ten_odd_set
````

**사전 조건제시법**

조건제시법을 이용하여 사전을 생성하는 과정도 유사하다. 
아래 코드는 0부터 10 사이의 홀수를 키로, 홀수의 제곱은 값으로 갖는 항목으로 구성된 사전을 생성한다.

````python
odd_squares = {x : x**2 for x in range(11) if x%2 == 1}
odd_squares
````

반면에 아래 코드는 문장에 포함된 단어를 키로, 단어의 길이를 값으로 갖는 항목들로 구성된 사전을 생성한다.

````python
words = '파이썬은 범용 프로그래밍 언어입니다.'.split()
print(words)
````

````python
len_dict = {k : len(k) for k in words}
len_dict
````
