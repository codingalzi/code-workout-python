**문제 2**

영어 단어는 키로, 단어의 뜻은 값으로 하는 사전이 다음과 같이 주어졌다.

````python
eng_dict = {'dog':'개', 
            'cat':'고양이',
            'lion':'사자',
            'tiger':'호랑이',
            'snake':'뱀'}
````

영어 단어가 인자로 입력되었을 때 사전의 키로 사용되었다면 뜻을 반환하고,
아니면 아래 문장을 반환하는 `eng_fun()` 함수를 선언하라.
단, 대소문자는 구분하지 않는다.

```
찾는 단어가 없습니다.
```

힌트: 문자열 `lower()` 메서드, 사전 `in` 연산자, 사전 인덱싱 활용

답:

사전에 키로 사용된 영어 단어는 모두 소문자만 사용한다.
따라서 입력값을 먼저 소문자만 사용하도록 변환한 다음에 키로 사용되었는지 여부를 판단한다.

````python
def eng_fun(word):
    word = word.lower()        # 대소문자 구분 없이 검색하기 위해 소문자로 변환
    if word in eng_dict:       # 단어가 딕셔너리에 존재하는지 확인
        return eng_dict[word]
    else:
        return '찾는 단어가 없습니다.'
````

````python
eng_fun('dog')
````

````python
eng_fun('fox')
````

다음과 같이 사전의 `get()` 메서드를 이용하여 `eng_func()` 함수를 선언할 수도 있다.

````python
def eng_fun(word):
    word = word.lower()
    return eng_dict.get(word, '찾는 단어가 없어요.')
````

````python
eng_fun('dog')
````

````python
eng_fun('fox')
````

**문제 3**

아래 `item_price` 에는 음료와 과자의 가격이 들어 있다. 

````python
item_price = "커피-1050원, 우유-870원, 밀크티-1300원, 새우과자-950원, 감자칩-1100원"
````

(1) 아래 사전 객체를 가리키는 변수 `price_dict`를 정의하는 코드를 작성하라.

```python
{'커피' : 1050, '우유' : 870, '밀크티' : 1300, '새우과자' : 950, '감자칩' : 1100}
```

힌트: 문자열 `split()` 메서드, 리스트 해체 활용

답:

`item_price` 문자열을 먼저 쉼표 기준으로 쪼개어 `물건-가격` 형식의 문자열로 구성된 리스트를 생성한다.

````python
item_price.split(',')
````

`for` 반복문을 실행하면서 리스트의 각 문자열 항목을 이번에는 하이픈 기준으로 쪼개어 물건과 가격을 분리한다.
단, `strip()` 메서드를 먼저 적용하여 문자열 양끝에 있는 공백을 제거한 다음에 쪼개는 것이 좋다.

````python
' 우유-870원'.strip().split('-')
````

위 설명을 종합하여 `price_dict` 변수를 다음과 같이 정의할 수 있다.

````python
price_dict = dict()

for it_pr in item_price.split(','):
    item, price = it_pr.strip().split('-')
    price_dict[item] = int(price[:-1])
````

````python
price_dict
````

(2) `price_dict`를 이용하여 지급 액수와 구매할 물품을 인자로 입력받았을 때 구매가격과 찾는 물품에 따라
다음과 같이 출력하는 `vending_machine()` 함수를 선언하라.

```
잔돈: 300원
부족 액수: 500원
찾는 물품: 없음
```

힌트: `if` 조건문, 사전 인덱싱, f-문자열 활용

답:

`vending_machine()` 함수는 두 개의 매개 변수를 사용한다.

- `money`: 지급 액수
- `item`: 찾는 물품

찾는 물품이 사전의 키로 사용되었는지 여부와, 물품 가격에 따른 조건문을 적용하여 함수의 본문을 정의한다.

````python
def vending_machine(money, item):
    if item in price_dict:                     # 물품이 딕셔너리에 존재하는지 확인
        price = price_dict[item]
        if price <= money:                     # 투입한 돈이 물품 가격보다 많거나 같은지 확인
            print(f"잔돈: {money  - price}원")
        else:
            print(f"부족 액수: {price - money}원")
    else:
        print("찾는 물품: 없음")        
````

````python
vending_machine(2000, '커피')
````

````python
vending_machine(1000, '커피')
````

````python
vending_machine(1000, '새우깡')
````

어느 유료 주차장의 주차 요금은 아래와 같으며, 1일 최대 주차 요금은 4만원으로 책정된다.

아래 `car_in_out` 변수는 일반차량에 대한 입출차 기록이 아래 형식으로 저장된 리스트를 가리킨다.

```
입출차 시간 차량번호네자리 입출차여부
```

(1) 차량 번호 네 자리가 주어졌을 해당 차량의 입출차 시간을 항목으로 갖는 사전을 반환하는 `car_parking_info()` 함수를 정의하라.
예를 들어, 1234 차량의 경우 다음 사전이 반환되어야 한다.

```python
{'IN': '07:30', 'OUT': '15:15'}
```

`car_in_out` 리스트의 각 항목에 대해 `split()` 메서드를 적용하면 각 항목이 세 개의 문자열로 쪼개지어 결국 길이가 3인 리스트가 생성된다.
예를 들어 0번 인덱스 항목을 쪼갠 결과는 다음과 같다.

````python
car_in_out[0].split()
````

쪼갠 결과를 이용해서 해당 차량 번호의 입차 시간, 출차 시건 정보를 사전으로 구성하면 된다.
예를 들어 아래 코드는 1234 차량의 입차 시간, 출차 시간 정보를 담은 사전을 생성한다.

````python
car_id = '1234'     

car_parking_in_out = {}                    # 딕셔너리를 이용하여 차량의 입출차 시간 저장
for record in car_in_out:
    time, id, status = record.split()      # 시간, 차량 번호, 입출차 상태를 각각 변수에 저장
    if id == car_id:                       # 차량 번호가 일치하는 경우에만 입출차 시간 저장
        car_parking_in_out[status] = time

print(car_parking_in_out)
````

위 코드를 함수의 본문으로 사용하면 `car_parking_info()` 함수를 정의할 수 있다.

- 함수 매개변수: `car_id`
- 함수 반환값: `car_parking_in_out`

````python
def car_parking_info(car_id):     

    car_parking_in_out = {}                    # 딕셔너리를 이용하여 차량의 입출차 시간 저장
    for record in car_in_out:
        time, id, status = record.split()      # 시간, 차량 번호, 입출차 상태를 각각 변수에 저장
        if id == car_id:                       # 차량 번호가 일치하는 경우에만 입출차 시간 저장
            car_parking_in_out[status] = time

    return car_parking_in_out
````

(2) 입차 시간과 출차 시간이 `'07:30'` 형식의 문자열로 주어졌을 때 분 단위의 주차 시간을 반환하는 `parking_duration()` 함수를 정의하라.

답:

예를 들어 입차 시간이 `'07:30'` 문자열로 주어졌을 때 몇 시 몇 분에 입차했는지를 먼저 확인하기 위해
문자열 `split()` 메서드를 활용한다.

````python
car_in = '07:30'
in_hour, in_min = car_in.split(':')
print(f"입차 시간: {in_hour}시 {in_min}분")
````

출차 시간에 동일한 기법을 적용한다.

````python
car_out = '13:15'
out_hour, out_min = car_out.split(':')
print(f"출차 시간: {out_hour}시 {out_min}분")
````

따라서 분 단위의 주차 시간은 다음과 같이 계산된다.

````python
duration = (int(out_hour) * 60 + int(out_min)) - (int(in_hour)*60 + int(in_min))
print(f"주차 시간: {duration}분")
````

위 설명을 종합하여 `parking_duration()` 함수를 다음과 같이 정의할 수 있다.

````python
def parking_duration(car_in, car_out):
    in_hour, in_min = car_in.split(':')     # 입차 시간에서 시와 분을 분리
    out_hour, out_min = car_out.split(':')  # 출차 시간에서 시와 분을 분리
    # 
    duration = (int(out_hour) * 60 + int(out_min)) - (int(in_hour)*60 + int(in_min))
    return duration
````

- `1234` 차량 주차 시간

````python
parking_duration('07:30', '15:15')
````

- `2580` 차량 주차 시간

````python
parking_duration('07:35', '08:45')
````

(3) 분 단위의 주차 시간이 주어졌을 때 주차 요금을 반환하는 `parking_fee()` 함수를 정의하라.

답:

주차 시간에 따라 주차 요금이 결정된다.

- 최초 30분: 2000원
- 이후 10분 단위: 500원
    - 10분 단위로 끊은 후 몇 분이 지나면 500원 추가
- 계산된 주차 요금이 4만원을 초과한 경우 4만원으로 최종 주차 요금 지정

`parking_fee()` 함수의 본문을 위 내용에 따라 `if` 조건문으로 구현하면 다음과 같다.

````python
def parking_fee(duration):
    if duration <= 30:                                # 30분 이내는 2000원
        fee = 2000
    elif (duration - 30) % 10 == 0:                   # 30분 초과 후 10분마다 500원 추가
        fee = 2000 + ((duration-30) // 10) * 500
    else:                                             # 10분 단위로 계산하므로 나머지가 있으면 1단위 추가
        fee = 2000 + ((duration-30) // 10 + 1) * 500
        
    fee = min(fee, 40000)                             # 최대 요금은 40000원
    return fee
````

- `1234` 차량 주차 요금

````python
parking_fee(465)
````

- `2580` 차량 주차 요금

````python
parking_fee(70)
````

(4) 차량 번호가 주어지면 주차 비용을 반환하는 `car_parking_fee()` 함수를 정의하라.

힌트: 이전 문제들의 함수 활용

답:

다음 순서를 따른다.

1. `car_parking_info()` 활용: 차량 입출차 시간 확인
2. `parking_duration()` 활용: 차량 입출차 시간 정보에서 분 단위 주차 시간 계산. 사전 인덱싱 활용.
3. `parking_fee()` 활용: 분 단위 주차 시간을 이용하여 주차 비용 계산

````python
def car_parking_fee(car_id):
    parking_info = car_parking_info(car_id)               # 차량의 입출차 시간 사전 생성
    if 'IN' in parking_info and 'OUT' in parking_info:    # 입차와 출차 시간이 모두 존재하는지 확인
        duration = parking_duration(parking_info['IN'], 
                                    parking_info['OUT'])  # 주차 시간 계산
        fee = parking_fee(duration)                       # 주차 요금 계산
        return fee
    else:                                                 # 입출차 정보가 완전하지 않은 경우
        return "입출차 정보가 완전하지 않습니다."
````

````python
car_parking_fee('1234')
````

**문제 2**

리스트를 내림차순으로 정렬한 리스트를 반환하는 함수 `sort_elem()`를 구현하라.
단, 중복 항목은 무시해야 한다.

예를 들어 아래와 같이 함수가 작동해야 한다.

```
sort_elem([2, 5, 2, 3, 3, 8, 2, 7]) = [8, 7, 5, 3, 2]
sort_elem([15, 3, 15, 1, 3]) = [15, 3, 1]
```

힌트: `sorted()` 함수 활용

답:

먼저 중복을 무시하도록 `set()` 함수를 적용한 다음에 다시 리스트로 변환한다.

````python
list(set([2, 5, 2, 3, 3, 8, 2, 7]))
````

**문제 3**

다음과 같이 주문 번호가 리스트로 주어졌다.

````python
orders = ['A102', 'A105', 'A102', 'A107', 'A105', 'A110', 'A102']
````

두 번 이상 등장한 주문 번호를 처음 중복이 확인된 순서대로 담은 리스트 `duplicated_orders`를 생성하는 코드를 작성하라.
코드 실행 결과는 다음과 같아야 한다.

```python
['A102', 'A105']
```

힌트: 

- `for` 반복문, `in` 연산자, 집합 `add()` 메서드, 리스트 `append()` 메서드 활용
- 집합 `add()` 메서드는 기존 집합에 항목을 추가함.

답:

`set()` 함수로 리스트 전체를 한 번에 변환하면 기존 항목들의 순서가 사라진다.
따라서 이미 확인된 주문, 두 번 이상 주문이 확인된 주문 등은 별도의 집합으로 따로 관리하면서
동시에 두 번 이상 주문된 번호들은 별도의 리스트에 추가한다.

````python
orders = ['A102', 'A105', 'A102', 'A107', 'A105', 'A110', 'A102']

# 확인된 주문 번호를 저장하기 위한 집합
seen = set()
# 중복 확인된 주문 번호를 저장하기 위한 집합
duplicated = set()
# 중복 주문된 번호를 순서대로 저장하기 위한 리스트
duplicated_orders = []

for order in orders:
    if order in seen and order not in duplicated: # 이미 본 주문이면서 아직 중복으로 기록되지 않은 경우
        duplicated_orders.append(order)           # 중복 주문 리스트에 추가
        duplicated.add(order)                     # 중복으로 기록된 주문으로 추가
    else:
        seen.add(order)                           # 주문을 본 적이 없으면 seen 집합에 추가

duplicated_orders
````

**문제 3: 상품 할인표 만들기**

문구점에서 다음 상품들의 원래 가격이 사전으로 주어졌다.

````python
prices = {"연필": 500,
          "공책": 1200,
          "지우개": 800,
          "필통": 5000}
````

사전 조건제시법을 이용하여 모든 상품 가격을 10% 할인한 새 사전을 생성하라.
단, 할인 가격은 정수로 저장한다.
실행 결과는 다음과 같아야 한다.

```python
{"연필": 450, "공책": 1080, "지우개": 720, "필통": 4500}
```

답:

사전 조건제시법은 `상품명: 할인가격` 형식의 항목을 생성한다.
사전의 `items()` 메서드를 이용하면 상품명과 가격을 함께 꺼낼 수 있다.
10% 할인 가격은 원래 가격에 `0.9`를 곱하여 계산하고, `int()` 함수로 정수로 변환한다.

예를 들어 `"연필"`의 가격은 `500`원이므로 할인 가격은 `int(500 * 0.9)`, 즉 `450`원이다.

````python
discounted_prices = {item: int(price * 0.9) for item, price in prices.items()}
discounted_prices
````
