### 연습문제

**문제 1**

두 변수가 다음과 같이 선언된다.

````python
x = 2.4
y = 5.7
````

두 변수가 가리키는 값의 조화 평균을 계산하는 표현식을 정의한 다음에 변수 z에 할당한다.
또한 z가 가리키는 값과 자료형을 화면에 다음과 같이 출력한다.

```python
2.4 와 5.7 의 조화평균: 3.38
```

단, `x`와 `y`의 조화평균은 다음과 같이 계산하며, 소수점 이하 셋째 자리에서 반올림한다.

```python
2 / (1/x + 1/y)
```

답:

두 변수 `x`와 `y`가 가리키는 값의 조화평균을 나타내는 표현식은 `2 / (1/x + 1/y)`이며,
`round()` 함수를 활용하여 지정된 방식으로 조화평균을 출력할 수 있다.

````python
# 조화 평균
mean_harmony = 2 / (1/x + 1/y)

print(x, "와", y, "의 조화평균:", round(mean_harmony, 2))
````

**문제 2**

대기실에 8개의 의자가 일렬로 놓여 있다.
대기실에 있는 사람들이 의자에 앉을 때 어느 두 사람도 이웃하여 못하도록 하면서 앉을 수 있는 최대 인원이 앉는 방식과
최소 인원이 앉는 방식을 문자열로 출력하는 코드를 작성하라.

단, 자리 앉기가 완료된 이후에는 더 이상 어느 누구도 자리에 앉을 수 없어야 한다.
예를 들어 아래 방식으로 앉는 것은 허용되지 않는다.
아래 표에서 `O`은 채워진 자리를, `X`는 빈자리를 가리킨다.

| 불가능 방식 | 이유 |
| :---: | :--- |
| `XXOXOXOX` | 왼쪽 끝에 한 명 더 앉을 수 있음 |
| `XOXXOXXX` | 오른쪽 끝에 한 명 더 앉을 수 있음|

답:

최대 인원 계산법:

8개의 의자가 있을 때, 예를 들어, 맨 왼쪽 의자에 앉은 다음에 한 자리 건널 때마다 앉으면 최대 인원이 앉게 된다.
즉 두 의자당 하나에 한 사람이 앉으면 된다. 
따라서 최대 인원은 4명이다.

```
OXOXOXOX
```

최소 인원 계산법:

최소 인원을 계산하기 위해 먼저 의자가 3개라고 가정하자. 
그러면 아래 방식으로 앉을 때 1명을 초과해서 앉을 수 없다.

```
XOX
```

이 점을 활용하면 최소 인원으로 앉는 방법을 구현할 수 있다.
예를 들어, 아래 언급된 방식처럼 `XOX` 형식을 두 번 반복한 다음에 나머지 하나 또는 두 자리 중에 한 자리에 앉는 방법이 최소인원이 된다.


```
XOXXOXXO
```

위 설명에 따르면 최대 인원수가 앉는 방식은 `OX`가 반복되는 것이고,
최소 인원수가 앉는 방식은 `XOX`가 반복되는 것이다.
따라서 8을 2 또는 3으로 나눌 때의 몫이 중요함을 알 수 있다.
이 성질을 이용하여 최대 인원이 앉는 방식과 최소 인원이 앉는 방식을
출력하는 코드를 다음과 같이 작성할 수 있다.

먼저, 의자 개수를 가리키는 변수 `chairs`를 정의한다.

````python
# 의자 개수
chairs = 8
````

`max_num`은 의자에 앉을 수 있는 최대 인원을 가리킨다.

````python
# 최대 인원
max_num = chairs//2
````

최대 인원이 앉는 방식은 다음과 같다.

````python
max_seats = "OX" * max_num
print(max_seats)
````

`min_num`은 의자에 앉을 수 있는 최소 인원을 가리킨다.

````python
# 최소 인원
min_num = (chairs//3) + 1
````

최소 인원이 앉는 방식은 다음과 같다.

````python
min_seats = "XOX" * (min_num-1) + "XO"
print(min_seats)
````

언급된 모든 코드를 종합하면 다음과 같다.

````python
# 최대 인원
max_num = chairs//2
max_seats = "OX" * max_num

# 최소 인원
min_num = (chairs//3) + 1
min_seats = "XOX" * (min_num-1) + "XO"


print("최대 인원", max_num, "명:", max_seats)
print("최소 인원", min_num, "명:", min_seats)
````

## 자료형

### 연습문제

**문제 1**

`dogs`와 `cats` 두 변수를 다음과 같이 선언한다.

````python
dogs = '12'
cats = '7'
````

강아지가 고양이보다 몇 마리 더 많은지를 아래 형식으로 
출력하는 코드를 작성하라.

```python
강아지가 고양이보다 5 마리 더 많다.
```

답:

두 변수가 가리키는 값이 모두 정수 형식의 문자열이다.
따라서 `int()` 함수를 두 변수에 적용하면 연산이 가능해지며,
강아지가 고양이보다 5 마리 더 많음을 계산할 수 있다.

````python
diff = int(dogs) - int(cats)
print("강아지가 고양이보다", diff, "마리 더 많다.")
````

**문제 2**

리스트가 다음과 같이 주어졌다. 

````python
data = [3, "py", 2.5, "7", True, "hi"]
````

변수 `item` 하나만 이용해서 아래 문자열을 
가리키는 변수 `dynamic_typing`를 정의하라.

단, 문자열 연산, 리스트 인덱싱, 형변환 함수를 모두 활용해야 하며,
`item` 변수를 한 번 정의한 다음에 5번 재할당 해야 한다.
또한 `dynamic_typing`는 빈 문자열을 가리키도록 한 다음에 5번 업데이트 해서
최종적으로 아래 문자열을 가리키도록 해야 한다.

```python
'3, py, 2, 7.0, 1, True'
```

답:

생성해야 하는 문자열은 리스트에 포함된 항목들과 일대일 대응된다.
다만 일부 항목이 변형된 형태로 연결될 뿐이다.
또한 변환된 값은 원래의 값에 적절한 형변환 함수를 적용하면 된다.

| 원래 값 | 변환된 값 | 형변환 함수 |
| :--- | :--- | :--- |
| 정수 `3` | 정수 `3` | 없음 |
| 문자열 `'py'` | 문자열 `'py'` | 없음 |
| 부동소수점 `2.5` | 정수 `2` | `int()` |
| 문자열 `"7"` | 부동소수점 `7.0` | `float()` |
| 부울값 `True` | 정수 `1` | `int()` |
| 문자열 `"hi"` | 부울값 `True` | `bool()` |


`dynamic_typing` 변수가 가리키는 값을 빈 문자열로 지정한 다음에
변수 업데이트 과정을 통해 지정된 문자열을 최종적으로 가리키도록 한다.
그 다음에 앞서 언급한 대로 `data` 변수가 가리키는 리스트의 항목을
추출해서 변환한 다음에 `dynamic_typing` 변수를 업데이트하는
과정을 반복한다.

````python
dynamic_typing = ""

item = data[0]
dynamic_typing += str(item) + ", "        # 정수 -> 문자열

item = data[1]
dynamic_typing += item + ", "             # 문자열 그대로

item = data[2]
dynamic_typing += str(int(item)) + ", "   # 실수 -> 정수 -> 문자열

item = data[3]
dynamic_typing += str(float(item)) + ", " # 문자열 -> 실수 -> 문자열

item = data[4]
dynamic_typing += str(int(item)) + ", "   # 부울값 -> 정수 -> 문자열

item = data[5]
dynamic_typing += str(bool(item))         # 문자열 -> 부울값 -> 문자열

print(dynamic_typing)
````
