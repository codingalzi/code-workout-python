## 숫자 맞히기 게임

### 연습문제

**문제 1**

아래 코드의 실행 결과를 설명하라.
반복문 실행횟수는 얼마인가?

````python
def counter_func(num, limit) :
    i = 1
    while True :
        print(i)
        if i == limit :
            return i
        else:
            print("=" * i)
            print("*" * i)
            i += 1
        return num

counter_func(100, 5)
````

**답**

`counter_func(100, 5)`이 호출되면 함수의 본문이 차례대로 실행된다.

- `i = 1` 에서 시작하고 `while True` 반복문이 실행된다. 
- 논리식 `True`가 참이라 `while` 반복문이 무한 반복되어야 하는 것처럼 보인다.
    - `print(i)`에 의해 1이 출력되고
    - `if ... else ...` 조건문에서 `i == limit`가 거짓이기에 `else`의 본문이 실행되어
    - `"="`와 `"*"` 가 출력된다.
    - 바로 이어서 `return 100` 이 실행되어 100을 반환하고 반복문을 비롯해서 함수의 실행이 완전히 종료된다.

결국 `while` 반복문은 1번만 실행되었다.

**문제 2**

임의로 생성된 숫자를 맞히는 게임을 실행하는 `guessing_number_game()` 함수를 정의하라.
단, 두 개의 매개변수 `start`와 `end`를 사용해야 하며, 
두 매개변수의 인자는 `random.randint()` 함수에게 전달되어
반환되는 임의의 숫자가 포함되는 구간으로 사용되도록 한다.

답:

````python
def guessing_number_game(start, end):
    import random

    print("숫자 맞히기 게임에 환영합니다.")

    # 맞혀야 하는 값을 start와 end 사이의 정수 중에서 하나를 무작위로 선택하도록 함.
    secret = random.randint(start, end)
    guess = -1   # 이어지는 while 반복문이 최소 한 번은 실행되도록 함

    while guess != secret:
        guess = int(input(f"{start}부터 {end} 사이의 정수 하나를 입력하세요: "))
        print(f"입력값: {guess}")

        if guess == secret:
            print("맞았습니다!")
        elif guess > secret:
            print("너무 커요!")
        else:
            print("너무 작아요!")

    print("게임 종료!")
````

아래 코드는 1과 10 사이의 정수를 맞히는 게임을 실행한다.

````python
guessing_number_game(1, 10)
````

반면에 아래 코드는 30과 50 사이의 정수를 맞히는 게임을 실행한다.

````python
guessing_number_game(30, 50)
````

## 이진 탐색

### 연습문제

**문제 1**

아래 코드의 실행 결과를 설명하라.
탐색 범위 `[low, high]`가 어떻게 변하는지 단계별로 설명하라.

````python
sorted_list = [1, 3, 5, 7, 9, 11, 13, 15]
target = 7

low = 0
high = len(sorted_list) - 1

while low <= high:
    mid = (low + high) // 2
    if sorted_list[mid] == target:
        print(f"{target}은(는) 인덱스 {mid}에 있습니다.")
        break
    elif sorted_list[mid] > target:
        high = mid - 1
    else:
        low = mid + 1
else:
    print("찾는 값이 없습니다.")
````

답:

- 초기 상태: `low=0`, `high=7`, `mid=3`, `sorted_list[3]=7` → `target`과 같으므로 즉시 찾음.
- 따라서 반복문은 **1번**만 실행되고 종료된다.
- 출력: `7은(는) 인덱스 3에 있습니다.`

리스트의 정중앙에 `target`이 위치한 경우 이진 탐색은 단 1번의 시도만으로 값을 찾아낸다.

**문제 2**

아래와 같이 오름차순으로 정렬된 리스트가 주어졌다.

```python
sorted_list = [2, 5, 8, 12, 16, 23, 38, 45, 67, 91]
```

이진 탐색 알고리즘을 이용하여 `target = 23`이 리스트에 존재하는지 확인하고,
존재한다면 그 인덱스를 출력하는 코드를 작성하라.
존재하지 않는다면 "찾는 값이 없습니다."를 출력하라.

단, `while` 반복문을 사용해야 하며, `low`와 `high` 변수를 활용하여
탐색 범위를 절반씩 줄여 나가야 한다.

답:

`sorted_list`에서 `target`을 이진 탐색으로 찾는다.

- `low`와 `high`는 탐색 범위의 시작과 끝 인덱스를 가리킨다.
- 매 반복마다 중간 인덱스 `mid = (low + high) // 2`를 계산하여 `sorted_list[mid]`와 `target`을 비교한다.
    - `sorted_list[mid] == target`: 찾았으므로 인덱스를 출력하고 `break`로 반복 종료.
    - `sorted_list[mid] > target`: `target`이 중간값보다 작으므로 `high = mid - 1`로 상한을 낮춤.
    - `sorted_list[mid] < target`: `target`이 중간값보다 크므로 `low = mid + 1`로 하한을 높임.
- `while ... else` 구문을 사용하여, 반복문이 `break` 없이 정상 종료되면(즉 `target`을 찾지 못하면) `else` 블록이 실행되어 "찾는 값이 없습니다."를 출력한다.

`target = 23`은 인덱스 5에 위치하므로 아래와 같이 출력된다.

```
23은(는) 인덱스 5에 있습니다.
```

````python
sorted_list = [2, 5, 8, 12, 16, 23, 38, 45, 67, 91]
target = 23

low = 0
high = len(sorted_list) - 1

while low <= high:
    mid = (low + high) // 2
    if sorted_list[mid] == target:
        print(f"{target}은(는) 인덱스 {mid}에 있습니다.")
        break
    elif sorted_list[mid] > target:
        high = mid - 1
    else:
        low = mid + 1
else:
    print("찾는 값이 없습니다.")
````

위 리스트에 포함된 값들 중에서 최댓값과 최댓값의 위치(인덱스)를 구하라.

답:

인덱스 변수로 리스트 전체를 대상으로

`max_idx`와 `max_val`은 각각 0과 리스트의 0번 인덱스 값으로 지정한다.

````python
numbers = [5, 3, 8, 1, 9, 2]

max_val = numbers[0]
max_idx = 0
i = 1
while i < len(numbers):
    if numbers[i] > max_val:
        max_val = numbers[i]
    i += 1

print("최댓값:", max_val)
````

**문제 2**

문자열이 다음과 같이 주어졌다.

````python
text = "racecar"
````

위 문자열이 회문인지 판단하는 코드를 작성하라.
단, 회문은 왼쪽에서부터 읽을 때와 오른쪽에서부터 읽을 때 동일한 단어를 가리킨다.

힌트 1: `while` 반복문과 문자열 인덱싱을 활용하세요.
힌트 2: 인덱스 `i`를 왼쪽부터 출발시키고, 역방향 문자의 인덱스를 `len(text) - 1 - i`로 두어 비교해 보세요. (서로 다른 두 변수 `left`, `right`를 두는 투 포인터 활용도 좋은 방법입니다)

답:

`i`는 문자열 앞에서, `j`는 뒤에서 시작해 서로 가까워지며 문자를 비교한다.
두 문자가 같으면 안쪽으로 이동하고, 다르면 즉시 반복을 멈추고
회문이 아니라고 출력한다.
회문인 경우엔 `i >= j`가 참이 순간에 반복문이 멈춘다.

````python
text = "racecar"

i = 0
j = len(text) - 1

while i < j and text[i] == text[j]:
    i += 1
    j -= 1

if i >= j:
    print(text, "는 회문입니다.")
else:
    print(text, "는 회문이 아닙니다.")
````

## continue와 break

### 연습문제

**문제 1**

아래 두 코드의 실행 결과가 동일한 이유를 설명하라.

- 코드 1

````python
for i in range(4):
    j = 0
    while j < 4:
        if i < j:
            break
        print([i, j])
        j += 1
````

- 코드 2

````python
i = 0
while i in range(4):
    for j in range(4):
        if i < j:
            continue
        print([i, j])
        
    i += 1
````

답:

코드 1은 다음과 같이 작동한다.

- `for` 반복문에 의해 `i`가 0부터 3까지 변함.
- 각각의 `i`에 대해 `i < j` 가 거짓인 경우, 즉 `j <= i`가 참인 경우에만 
    리스트 `[i, j]` 출력하고 `j`를 1만큼 증가시키고 `while` 명령문 반복.
- 각각의 `i`에 대해 `i < j` 가 참이 되는 순간 `break` 명령문이 실행되어
    `while` 반복문이 종료되어, `for` 반복문에서 `i`의 다음 경우로 넘어감.
- 결론적으로 `i >= j`인 경우에만 `[i, j]` 출력

코드 2는 다음과 같이 작동한다.

- `while` 반복문에 의해 `i`가 0부터 3까지 변함.
- 각각의 `i`에 대해 `i < j` 가 거짓인 경우에만 리스트 `[i, j]` 출력하고 `for` 반복문에서 `j`의 다음 경우로 넘어감..
- 각각의 `i`에 대해 `i < j` 가 참이면 `continue` 명령문이 실행되어 `for` 반복문에서 `j`의 다음 경우로 넘어감..
- 결론적으로 `i >= j`인 경우에만 `[i, j]` 출력.

**문제 2**

아래 코드의 실행결과를 설명하라.

````python
n = 3
while n < 10:
    is_even = True

    if n % 2 == 1:
        is_even = False
        print("-", n, "은(는) 홀수다.")
    
    break

    if is_even:
        print("-", n, "은(는) 짝수다.")
        
    n += 1
    
print("실행 종료!")
````

답:

`while` 반복문은 `n`이 3부터 10까지의 정수를 차례대로 가리키는 동안 실행된다.

- 5번 줄에서 시작하는 `if` 조건문은 `n`이 홀수일 때 홀수임을 확인한다.
- 이후에 9번 줄에서 `break` 명령문이 실행되어 `while` 반복문의 실행이 중단된다.
- 결국 `n`이 3인 경우에 홀수임을 확인한 후에 곧바로 `while` 반복문의 실행이 중단된다.
- `while` 반복문이 종료된 후에 이어지는 명령문이 계속해서 실행되어
    마지막 줄의 `print("실행 종료!")`에 의해 코드의 실행이 종료되었음을 확인해준다.

**문제 3**

아래 코드는 1과 100 사이에서 임의로 지정된 `secret`를
맞힐 때까지 코드를 실행한다.

````python
import random

secret = random.randint(1, 100)
low = 1
high = 100
count = 0
guess = -1  # secret은 1~100이므로 -1로 초기화하면 반복문이 반드시 실행됨

while low <= high and guess != secret:
    guess = (low + high) // 2
    print(f"low: {low}, high: {high}", end="  →  ")
    count += 1
    print(f"시도 {count}: {guess}", end="  →  ")

    if guess == secret:
        print("맞았습니다!")
    elif guess > secret:
        print("너무 커요!")
        high = guess - 1
    else:
        print("너무 작아요!")
        low = guess + 1

print(f"{count}번 만에 찾았습니다.")
````

위 코드에서 반복문의 실행은 아래 조건식에 의해 결정된다.

```python
low <= high and guess != secret
```

위 조건식 대신에 `low <= high`를 조건식으로 사용하는 
`while` 반복문으로 동일한 기능을 수행하는 코드를 작성하라.

힌트: `break` 명령문 활용

답:

조건식 `low <= high and guess != secret`에서
`guess != secret`는 `guess`와 `secret` 두 변수가 가리키는
값이 다른 동안 반복문을 실행하도록 설정하기 위해 사용되었다.

이는 반복문이 실행될 때 두 변수가 가리키는 값이 같아지는 순간에 반복문의
실행이 멈춤을 의미한다.
따라서 `if guess == secret` 조건문의 본문에
`break` 명령문을 추가하면 된다.

````python
import random

secret = random.randint(1, 100)
low = 1
high = 100
count = 0

while low <= high:
    guess = (low + high) // 2
    print(f"low: {low}, high: {high}", end="  →  ")
    count += 1
    print(f"시도 {count}: {guess}", end="  →  ")

    if guess == secret:
        print("맞았습니다!")
        break
    elif guess > secret:
        print("너무 커요!")
        high = guess - 1       # 빈칸 1
    else:
        print("너무 작아요!")
        low = guess + 1        # 빈칸 2

print(f"{count}번 만에 찾았습니다.")
````
