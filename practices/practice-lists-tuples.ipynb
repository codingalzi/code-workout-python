**문제 2**

사용자의 이름과 전화번호를 입력 받으면 다음과 같이 반환하는 
`phone_book()` 함수를 구현하라.
단, 사용자의 이름은 두 글자에서 네 글자 사이며, 
핸드폰 번호는 9개에서 11개의 숫자가 다양한 방식으로 주어진다.

```
phone_book('김강현', '010-1234-5678') = ('김*현', 5678)
phone_book('강현', '01012345678') = ('강*', 5678)
phone_book('김강현이', '031-123-5678') = ('김*현이', 5678)
phone_book('강현이', '02-123-5678') = ('강*이', 5678)
phone_book('김강현', '010.1234.5678') = ('김*현', 5678)
```

힌트: 리스트 변환, 문자열 `join()` 메서드 활용

답:

아래 `phone_book()` 함수는 지정된 방식으로 추출된 이름과 전화번호 네자리를 반환한다.

- 이름: 리스트로 변환한 후에 둘째 글자를 `*`로 대체 후 다시 문자열로 변환
- 전화번호: 슬라이싱으로 끝에서 네자리 추출

````python
def phone_book(name, phone):
    # 이름 변환
    letter_list = list(name)             # 이름을 글자 리스트로 변환
    letter_list[1] = "*"                 # 이름의 2번째 글자를 "*"로 수정
    censored_name = "".join(letter_list) # 수정된 글자 리스트를 다시 문자열로 합침
    
    # 전화번호 끝에서 네자리 추출
    censored_phone = int(phone[-4:])     # 전화번호의 마지막 4자리 숫자만 추출하여 정수로 변환

    return (censored_name, censored_phone)
````

````python
print(phone_book('김강현', '010-1234-5678'))
print(phone_book('강현', '01012345678'))
print(phone_book('김강현이', '031-123-5678'))
print(phone_book('강현이', '02-123-5678'))
print(phone_book('김강현', '010.1234.5678'))
````

**문제 3**

직사각형 모양의 바닥에 정사각형 모양의 타일을 깔아야 한다.
타일은 온장을 그대로 사용할 수도 있고, 잘라서 일부분만 사용할 수도 있다.
잘라서 사용한 타일의 나머지는 버린다.

공간의 가로와 세로 크기, 그리고 타일의 크기가 cm 단위로 주어졌을 때
타일을 깔 때 필요한 온장 타일과 잘라서 사용한 타일의 개수를 반환하는 함수 `tiling_tuple()`을 구현하라.
단 공간 크기의 가로, 세로, 타일의 크기는 정수라고 가정하며,
타일의 크기의 기본값은 한 변의 길이가 30cm인 정사각형이다.

예를 들어, 가로와 세로 각각 4미터, 3미터인 공간을 한 변의 길이가 30cm인 타일과 35cm 타일로 채우는 경우 각각에 대해
필요한 타일의 개수를 계산하라.

답:

`tiling()` 함수의 매개변수는 다음과 같이 한다.

- `width`: 공간의 가로 크기. 단위는 cm.
- `height`: 공간의 세로 크기. 단위는 cm.
- `tile_length=30`: 타일의 한 변의 크기. 단위는 cm. 기본 키워드 인자는 30.

````python
def tiling_tuple(width, height, tile_length=30):
    # 가로와 세로에 타일이 몇 개씩 들어가는지 계산
    num_full_width = width // tile_length
    num_full_height = height // tile_length

    # 전체 온장 타일 수 계산
    num_full = num_full_width * num_full_height

    # 부분 타일 수 계산
    if width % tile_length == 0:
        num_part_height = 0
        if height % tile_length == 0:
            num_part = 0
        else:
            num_part = num_full_width

    else:
        if height % tile_length == 0:
            num_part = num_full_height
        else:
            num_part = num_full_width + num_full_height + 1

    # 온장 타일 수와 부분 타일 수를 튜플로 반환
    return num_full, num_part
````

- 한 변의 길이가 30cm인 정사각형 타일을 사용한 경우

````python
tiling_tuple(400, 300)
````

- 한 변의 길이가 35cm인 정사각형 타일을 사용한 경우

````python
tiling_tuple(400, 300, 35)
````

````python
even_words = []                     # 짝수 인덱스 단어를 저장할 리스트 초기화

for idx, word in enumerate(words):  # words의 항목과 그 인덱스를 동시에 반복. idx는 인덱스, word는 단어
    if idx % 2 == 0:                # 인덱스가 짝수인 경우
        even_words.append(word)     # 짝수 인덱스 단어를 even_words에 추가

even_words
````

**문제 2**

두 개의 리스트가 다음과 같다.

````python
xs = [1, 3, 5, 7]
ys = [2, 4, 6, 8]
````

두 리스트의 같은 위치에 있는 항목끼리 더해 다음 리스트를 생성하라.

```python
[3, 7, 11, 15]
```

힌트: `zip()` 함수와 `append()` 메서드 활용

답:

`zip()` 함수로 두 리스트를 항목별로 묶은 후 동시 반복을 실행하면 된다.

````python
result = []               # 결과를 저장할 리스트 초기화

for x, y in zip(xs, ys):  # xs와 ys의 항목을 하나씩 묶어서 반복
    result.append(x + y)  # x와 y의 합을 result에 추가

result
````

**문제 3**

아래 코드는 인덱스와 리스트의 항목의 순서를 바꾼 튜플의 
리스트를 생성한다.

````python
some_list = ['foo', 'bar', 'baz', 'pyt', 'thon']

mapping = []
for idx, item in enumerate(some_list):
    mapping.append((item, idx))

mapping
````

`enumerate()` 함수를 사용하지 않으면서 동일한 기능을 구현하는 `for` 반복문을 작성하라.

답:

인덱스 정보가 필요하기에 `range()` 함수를 이용한다.

````python
some_list = ['foo', 'bar', 'baz', 'pyt', 'thon']

mapping = []
for idx in range(len(some_list)):
    mapping.append((some_list[idx], idx))

mapping
````

**문제 4**

메뉴와 가격이 공백으로 구분된 두 개의 문자열로 주어졌다.

````python
menu = "ham bread chicken egg"
prices = "1200 5000 17000 500"
````

(1) 두 문자열을 공백 기준으로 쪼개서 생성되는 두 리스트를 각각 `menu_list` 변수와 `price_list` 변수에 할당하라.

답:

문자열 `split()` 메서드를 활용한다.

````python
menu_list = menu.split()
price_list = prices.split()
````

(2) 가장 저렴한 메뉴와 가장 비싼 메뉴를 추천하는 문자열을
다음과 같이 출력하는 프로그램을 구현하라.

```python
가장 저렴한 메뉴: egg  
가장 비싼 메뉴: chicken  
```

힌트: `zip()` 함수, 리스트 `sort()` 메서드 활용

답:

먼저 메뉴와 가격을 맞춰서 튜플로 구성된 리스트를 생성한다.

````python
menu_prices = list(zip(menu_list, price_list))
menu_prices
````

가장 저렴한 메뉴를 찾기 위해 튜플을 가격 기준으로 정렬한다.

````python
menu_prices.sort(key=lambda x: int(x[1]))
menu_prices
````

가격 기준으로 정렬된 리스트의 0번 인덱스를 확인하면 최저가 메뉴와 가격을 출력할 수 있다.

````python
print(f"가장 저렴한 메뉴: {menu_prices[0][0]}")
print(f"가장 비싼 메뉴: {menu_prices[-1][0]}")
````

모든 코드를 정리하면 다음과 같다.

````python
menu_prices = list(zip(menu_list, price_list))
menu_prices.sort(key=lambda x: int(x[1]))
print(f"가장 저렴한 메뉴: {menu_prices[0][0]}")
print(f"가장 비싼 메뉴: {menu_prices[-1][0]}")
````
