### 연습문제

**문제 1**

다음 두 클래스에서 중복되는 인스턴스 변수와 메서드를 각각 찾아라.

```python
class Student:
    def __init__(self, name, age, school):
        self.name = name
        self.age = age
        self.school = school

    def introduce(self):
        print(self.name, self.age)


class Teacher:
    def __init__(self, name, age, subject):
        self.name = name
        self.age = age
        self.subject = subject

    def introduce(self):
        print(self.name, self.age)
```

답:

두 클래스에서 중복되는 인스턴스 변수는 `self.name`과 `self.age`이다.
두 생성자 모두 전달받은 이름과 나이를 같은 방식으로 저장한다.
또한 `introduce()` 메서드도 이름과 나이를 출력하는 동일한 기능을 구현한다.

`self.school`은 `Student`에만 있고 `self.subject`는 `Teacher`에만 있으므로 중복이 아니다.

**문제 2**

문제 1의 두 클래스를 그대로 유지한 채 `introduce()` 메서드가 이름 앞에 `"이름:"`을 출력하도록 수정하려 한다.
몇 곳을 수정해야 하는지 답하고, 이런 구조가 유지보수에 불리한 이유를 설명하라.

답:

`Student.introduce()`와 `Teacher.introduce()` 두 곳을 모두 수정해야 한다.
같은 기능이 여러 클래스에 중복되어 있으면 출력 형식이 바뀔 때마다 모든 구현을 찾아 수정해야 한다.
한 곳을 빠뜨리면 같은 역할의 메서드가 클래스마다 다르게 작동할 수 있으므로 유지보수에 불리하다.

**문제 3**

자동차를 나타내는 `Car` 클래스와 버스를 나타내는 `Bus` 클래스를 상속 없이 각각 정의하라.
두 클래스는 모두 `model`, `speed` 속성과 `move()` 메서드를 가져야 하며,
`Bus` 클래스에는 `capacity` 속성을 추가하라.
완성한 코드에서 어떤 부분이 중복되는지 주석으로 표시하라.

답:

상속을 사용하지 않으면 다음처럼 두 클래스에 공통 속성과 메서드를 각각 작성해야 한다.

````python
class Car:
    def __init__(self, model, speed):
        self.model = model  # Bus와 중복
        self.speed = speed  # Bus와 중복

    def move(self):  # Bus와 중복
        print(f"{self.model}이(가) 시속 {self.speed}km로 이동합니다.")


class Bus:
    def __init__(self, model, speed, capacity):
        self.model = model  # Car와 중복
        self.speed = speed  # Car와 중복
        self.capacity = capacity

    def move(self):  # Car와 중복
        print(f"{self.model}이(가) 시속 {self.speed}km로 이동합니다.")


car = Car("승용차", 100)
bus = Bus("시내버스", 60, 40)

car.move()
bus.move()
print("버스 정원:", bus.capacity)
````

`model`, `speed`를 저장하는 코드와 `move()` 메서드가 중복된다.
이 공통 부분을 부모 클래스로 옮기면 두 클래스가 같은 구현을 재사용할 수 있다.

## 상속

### 연습문제

**문제 1**

`Person`을 상속하는 `Student` 클래스를 정의하라.
`Student` 클래스는 `school` 속성을 추가로 가지며,
생성자에서 `super()`를 사용하여 이름, 나이, 성별을 초기화해야 한다.

답:

먼저 이름, 나이, 성별을 관리하는 `Person`을 정의한다.
`Student`는 `Person`을 상속하고, `super().__init__()`으로 부모 클래스의 초기화 기능을 재사용한다.

````python
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def introduction(self):
        print(f"이름: {self.name}")
        print(f"나이: {self.age}")
        print(f"성별: {self.gender}")


class Student(Person):
    def __init__(self, name, age, gender, school):
        super().__init__(name, age, gender)
        self.school = school
````

**문제 2**

문제 1의 `Student` 클래스에 `introduction()` 메서드를 재정의하라.
먼저 부모 클래스의 `introduction()`을 호출한 다음 학교 이름을 출력해야 한다.

답:

`super().introduction()`을 호출하면 이름, 나이, 성별을 출력하는 부모 클래스의 코드를 반복해서 작성하지 않아도 된다.
그다음 학생에게만 필요한 학교 정보를 출력한다.

````python
class Student(Person):
    def __init__(self, name, age, gender, school):
        super().__init__(name, age, gender)
        self.school = school

    def introduction(self):
        super().introduction()
        print(f"학교: {self.school}")

student = Student("김하나", 18, "여성", "파이썬고")
student.introduction()
````

**문제 4**

`Hero`를 상속하는 마법사 클래스 `MagicHero` 클래스를 정의하라.
`mana` 속성을 추가하고, 대상의 체력을 `mana`만큼 감소시키는 `magic_attack(target)` 메서드를 구현하라.
부모 클래스의 속성 초기화에는 `super()`를 사용한다.

답:

`MagicHero`의 생성자는 먼저 `Hero`의 생성자를 호출한 다음 `mana` 속성을 추가한다.
`magic_attack()`은 대상 객체의 `health`를 현재 객체의 `mana`만큼 감소시킨다.

````python
class Hero(Person):
    def __init__(
        self, name, age, gender, power, health, damage, inventory
    ):
        super().__init__(name, age, gender)
        self.power = power
        self.health = health
        self.damage = damage
        self.inventory = inventory

    def attack(self, target):
        target.health -= self.damage

# 마법사 클래스는 Hero 클래스를 상속받아 마나 속성을 추가하고, 마법 공격 메서드를 구현
class MagicHero(Hero):
    def __init__(
        self, name, age, gender, power, health, damage, inventory, mana
    ):
        super().__init__(
            name, age, gender, power, health, damage, inventory
        )
        self.mana = mana

    def magic_attack(self, target):
        target.health -= self.mana

# 마법사와 괴물 객체를 생성하고, 마법사가 괴물에게 마법 공격을 하는 예시
wizard = MagicHero(
    "마법사", 30, "여성",
    power=80,
    health=100,
    damage=20,
    inventory={"weapon": "지팡이"},
    mana=35,
)

monster = Hero(
    "괴물", 100, "미정",
    power=50,
    health=120,
    damage=15,
    inventory={},
)

wizard.magic_attack(monster)
print("괴물 체력:", monster.health)
````

## 구성

### 연습문제

**문제 1**

`Flying`을 상속하는 `TeleportFlying` 구상 클래스를 정의하라.
`fly()` 메서드는 `"순간이동합니다!"`를 출력해야 한다.
완성한 객체를 `thor.set_flying()`에 전달하여 동작을 확인하라.

답:

`Flying` 추상 클래스의 `fly()` 메서드를 구현하는 `TeleportFlying` 클래스를 정의한다.
완성된 객체를 `set_flying()`에 전달하면 영웅의 클래스를 바꾸지 않고 비행 동작을 교체할 수 있다.

````python
from abc import ABC, abstractmethod


class Flying(ABC):
    @abstractmethod
    def fly(self):
        pass


class SelfFlying(Flying):
    def fly(self):
        print("스스로 날아요!")


class TeleportFlying(Flying):
    def fly(self):
        print("순간이동합니다!")


class Hero:
    def __init__(self, name, flying=None):
        self.name = name
        self.flying = flying

    def fly(self):
        self.flying.fly()

    def set_flying(self, flying):
        self.flying = flying


thor = Hero("토르", SelfFlying())
thor.fly()
thor.set_flying(TeleportFlying())
thor.fly()
````

**문제 2**

비행 능력이 지정되지 않은 영웅이 `fly()`를 호출하면 `"날 수 없습니다."`를 출력하도록
`Hero.fly()` 메서드를 수정하라.

답:

`flying` 속성이 `None`인지 먼저 확인한다.
비행 객체가 있으면 기능을 위임하고, 없으면 안내 메시지를 출력한다.

````python
class Hero:
    def __init__(self, name, flying=None):
        self.name = name
        self.flying = flying

    def fly(self):
        if self.flying is None:
            print("날 수 없습니다.")
        else:
            self.flying.fly()

    def set_flying(self, flying):
        self.flying = flying


hulk = Hero("헐크")
thor = Hero("토르", SelfFlying())

hulk.fly()
thor.fly()
````

**문제 3**

공격 동작을 나타내는 `Attacking` 추상 클래스와 `PunchAttack`, `LaserAttack` 구상 클래스를 정의하라.
각 구상 클래스는 `attack()` 메서드에서 서로 다른 메시지를 출력해야 한다.

답:

`Attacking`은 모든 공격 방식이 제공해야 하는 `attack()` 메서드를 추상 메서드로 선언한다.
각 구상 클래스는 이 메서드를 자신에게 맞는 메시지로 구현한다.

````python
class Attacking(ABC):
    @abstractmethod
    def attack(self):
        pass


class PunchAttack(Attacking):
    def attack(self):
        print("주먹으로 공격합니다!")


class LaserAttack(Attacking):
    def attack(self):
        print("레이저로 공격합니다!")


PunchAttack().attack()
LaserAttack().attack()
````

**문제 4**

문제 3의 공격 객체를 `Hero`의 속성으로 저장하도록 클래스를 확장하라.
`set_attacking()` 메서드로 공격 방식을 실행 중에 교체할 수 있어야 한다.
이 설계에서 상속보다 구성이 유리한 점을 설명하라.

답:

`Hero`는 공격 객체를 `attacking` 속성으로 저장하고 실제 공격은 그 객체에 위임한다.
`set_attacking()`은 새로운 공격 객체를 저장하여 실행 중에 공격 방식을 교체한다.

````python
class Hero:
    def __init__(self, name, flying=None, attacking=None):
        self.name = name
        self.flying = flying
        self.attacking = attacking

    def fly(self):
        if self.flying is None:
            print("날 수 없습니다.")
        else:
            self.flying.fly()

    def set_flying(self, flying):
        self.flying = flying

    def attack(self):
        if self.attacking is None:
            print("공격할 수 없습니다.")
        else:
            self.attacking.attack()

    def set_attacking(self, attacking):
        self.attacking = attacking


ironman = Hero(
    "아이언맨",
    flying=SelfFlying(),
    attacking=LaserAttack(),
)
ironman.attack()

ironman.set_attacking(PunchAttack())
ironman.attack()
````

상속으로 공격 방식을 표현하면 공격 방식마다 새로운 영웅 자식 클래스가 필요하다.
구성을 사용하면 같은 `Hero` 객체에 저장된 공격 객체만 교체할 수 있다.
따라서 새로운 공격 방식을 추가하거나 실행 중에 동작을 변경하기가 더 쉽다.

**문제 3**

`Hero.set_flying()` 메서드가 `Flying`의 인스턴스만 받도록 검사 기능을 추가하라.
잘못된 객체가 전달되면 `TypeError`를 발생시켜야 한다.

```python
def set_flying(self, flying):
    # 여기에 검사 코드 추가
    self.flying = flying
```

답:

`isinstance()`로 전달된 값이 `Flying`의 인스턴스인지 확인한다.
검사를 통과하지 못하면 `TypeError`를 발생시키고, 통과한 경우에만 속성에 저장한다.

````python
class Hero:
    def __init__(self, name, flying=None):
        self.name = name
        self.flying = None
        if flying is not None:
            self.set_flying(flying)

    def fly(self):
        if self.flying is None:
            print("날 수 없습니다.")
        else:
            self.flying.fly()

    def set_flying(self, flying):
        if not isinstance(flying, Flying):
            raise TypeError("flying은 Flying의 인스턴스여야 합니다.")
        self.flying = flying


thor = Hero("토르", SelfFlying())
thor.fly()

try:
    thor.set_flying("로켓 비행")
except TypeError as error:
    print(type(error).__name__, error)
````

**문제 8: 신뢰를 이용한 협동 공격**

def team_attack(self, other, villain):

**문제 9: 최종 텍스트 게임 완성**

지금까지 구현한 내용을 하나의 프로그램으로 완성하라.

2. 모든 캐릭터가 자기소개한다.
3. 타이탄이 영웅들을 위협하고 헐크를 공격한다.

답:

아래 프로그램에서 각 객체는 자신의 상태를 관리하면서 다른 객체의 공개 메서드를 통해 협력한다.

대화와 행동으로 변화한 신뢰도가 마지막 협동 공격의 성공 여부를 결정한다.

````python
class Character(Human):
    def __init__(self, name, age, gender, health, damage):
        super().__init__(name, age, gender)
        self.health = health
        self.damage = damage

    def is_alive(self):
        return self.health > 0

    def take_damage(self, amount):
        self.health = max(0, self.health - amount)
        print(f"{self.name}: {amount}의 피해를 받았습니다. 남은 체력: {self.health}")


class Hero(Character):
    def __init__(self, name, age, gender, health, damage, power, flying=None):
        super().__init__(name, age, gender, health, damage)
        self.power = power
        self.trust = 0
        self.flying = flying

    def introduction(self):
        print(f"[영웅] {self.name} | 체력: {self.health} | "
              f"능력치: {self.power} | 신뢰도: {self.trust}")

    def talk(self, other, message):
        print(f"{self.name} -> {other.name}: {message}")

    def encourage(self, other):
        print(f"{self.name}: {other.name}, 네 힘이 필요해!")
        other.health = min(100, other.health + 10)
        other.trust += 1
        print(f"{other.name}의 체력: {other.health}, 신뢰도: {other.trust}")

    def respond(self):
        if self.trust == 0:
            message = "아직 혼자 싸우는 편이 낫겠어."
        elif self.trust < 3:
            message = "조금은 믿어도 될 것 같아."
        else:
            message = "우리는 한 팀이야."
        print(f"{self.name}: {message}")

    def rescue(self, other):
        if self.flying is None:
            print(f"{self.name}: 날 수 없어 {other.name}을 구조할 수 없습니다.")
            return False
        print(f"{self.name}이 {other.name}을 구조하러 출발합니다.")
        self.flying.fly()
        self.trust += 1
        other.trust += 1
        print(f"{self.name}과 {other.name}의 신뢰도가 증가했습니다.")
        return True

    def disagree(self, other):
        print(f"{self.name}과 {other.name}이 작전 방식을 두고 다툽니다.")
        self.trust = max(0, self.trust - 1)
        other.trust = max(0, other.trust - 1)

    def reconcile(self, other):
        print(f"{self.name}: 혼자 결정해서 미안해.")
        print(f"{other.name}: 다시 함께 싸우자.")
        self.trust += 1
        other.trust += 1

    def team_attack(self, other, villain):
        if self.trust < 2 or other.trust < 2:
            print(f"{self.name}과 {other.name}의 신뢰가 부족합니다.")
            return False
        attack_power = (self.power + other.power) // 2
        print(f"{self.name}과 {other.name}이 협동 기술을 사용합니다!")
        print(f"협동 공격력: {attack_power}")
        villain.take_damage(attack_power)
        return True


class TechHero(Hero):
    def specialty(self):
        print(f"{self.name}: 첨단 장비로 상황을 분석합니다.")


class MagicHero(Hero):
    def specialty(self):
        print(f"{self.name}: 번개의 힘을 불러옵니다.")


class PowerHero(Hero):
    def specialty(self):
        print(f"{self.name}: 강한 힘으로 장애물을 부숩니다.")


class Villain(Character):
    def __init__(self, name, age, gender, health, damage, goal):
        super().__init__(name, age, gender, health, damage)
        self.goal = goal

    def introduction(self):
        print(f"[악당] {self.name} | 체력: {self.health} | 목적: {self.goal}")

    def threaten(self, heroes):
        names = ", ".join(hero.name for hero in heroes)
        print(f"{self.name}: {names}! 너희는 나를 막을 수 없다!")

    def attack(self, other):
        print(f"{self.name}이 {other.name}을 공격합니다!")
        other.take_damage(self.damage)


ironman = TechHero("아이언맨", 45, "남성", 100, 20, 90, RocketFlying())
thor = MagicHero("토르", 1500, "남성", 100, 25, 95, SelfFlying())
hulk = PowerHero("헐크", 42, "남성", 100, 30, 100)
titan = Villain("타이탄", 1000, "미정", 90, 25, "도시의 에너지를 차지한다")
heroes = [ironman, thor, hulk]

print("=== 도시 수호대: 마지막 임무 ===")
print("\n[등장인물]")
for character in heroes + [titan]:
    character.introduction()

print("\n[영웅의 능력]")
for hero in heroes:
    hero.specialty()

print("\n[적의 등장]")
titan.threaten(heroes)
titan.attack(hulk)

print("\n[작전 회의]")
ironman.talk(thor, "내 계획대로 움직여 줘.")
thor.talk(ironman, "혼자 결정하려 하지 마라.")
ironman.disagree(thor)

print("\n[동료 구조]")
thor.encourage(hulk)
ironman.rescue(hulk)

print("\n[화해]")
ironman.reconcile(thor)
ironman.encourage(thor)
thor.encourage(ironman)

print("\n[현재 상태]")
for hero in heroes:
    print(f"{hero.name} | 체력: {hero.health} | 신뢰도: {hero.trust}")
    hero.respond()

print("\n[최종 협동 공격]")
ironman.team_attack(thor, titan)

print("\n[임무 결과]")
if titan.is_alive():
    print(f"타이탄의 남은 체력은 {titan.health}입니다. 임무를 계속합니다.")
else:
    print("영웅들이 힘을 합쳐 도시를 지켜냈습니다!")

print("=== 임무 종료 ===")
````
