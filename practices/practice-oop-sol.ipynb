**문제 1**

`polygon(t, length, n)` 함수는 한 변의 길이가 `length`인 정 n-각형을 그린다.
단, `t`는 거북이 객체를 가리킨다.

```python
def polygon(t, length, n):
    turn = 360/n  # 회전 각도
    for _ in range(n):
        t.forward(length)
        t.left(turn)
```

`polygon()` 함수를 이용하여 원을 그리는 `circle(t, r, n)` 함수를 구현하여라.
세 매개 변수 각각에 입력되는 값의 의미는 다음과 같다.

- `t`: 거북이 객체
- `r`: 원의 반지름
- `n`: 정 n-각형을 활용한 원 그리기.

예를 들어 `circle(t, 100, n=100)`를 실행하면 아래 모양의 원이 그려져야 한다.

<p><div align="center" border="1px"><img src="https://raw.githubusercontent.com/codingalzi/42H/master/jupyter-book/images/turtle16-1.png" width="300"/></div></p>

힌트: `n`각형을 그릴 때 한 변의 길이를 원 둘레를 이용하여 계산한다.

답:

원을 정 n-각형을 이용하여 근사적으로 그릴 수 있다. 단, n이 매우 커야 한다. 
또한 매우 큰 n에 대해 정 n-각형의 한 변의 길이가 `length`일 때 다음 식이 성립한다.

    원둘레 = 2 * math.pi * r = length * n
    
따라서 반지름이 `r`인 원을 정 n-각형으로 구현하려면
한 변의 길이가 다음과 같아야 한다.

    length = (2 * math.pi * r)/n

아래 `circle()` 함수는 원을 100각형으로 그린다.
반지름이 클 때는 `n`을 보다 키워줄 필요가 있다.

```python
def circle(t, r, n=100):
    length = (2 * math.pi * r) / n
    polygon(t, length, n)
```

**문제 2**

호는 부채꼴 모양으로 잘린 원의 일부를 가리킨다.

`polygon()` 함수를 수정하여 정 n-각형을 지정된 `angle` 만큼만 그리는 `polygon_arc(t, length, n, angle)` 함수를 정의하여라.
네 매개 변수 각각에 입력되는 값의 의미는 다음과 같다.

- `t`: 거북이 객체
- `length`: 다각형에 사용되는 한 변의 길이
- `n`: n-각형 지정
- `angle`: 호의 내각.

예를 들어 아래 그림은 한 변의 길이가 40인 16각형을 그릴 때 120도 만큼만, 즉 1/3 정도만 그린 결과를 보여준다.

```python
polygon_arc(t, 40, 16, 120)
```

<p><div align="center" border="1px"><img src="https://raw.githubusercontent.com/codingalzi/42H/master/jupyter-book/images/turtle12-1.png" width="300"/></div></p>

힌트: `polygon()` 함수의 본문에 사용된 `for` 반복문의 반복횟수를 `angle`에 의존하도록 한다.

답:

원을 한 바퀴가 360도이기에 `angle` 만큼 돌기 위해서는 아래 횟수만큼 회전한다.

`n * angle / 360`

```python
def polygon_arc(t, length, n, angle=360):
    turn = 360/n                  # 회전 각도
    n_iter = int(n * angle / 360) # 이동 반복 횟수
    for _ in range(n_iter):
        t.forward(length)
        t.left(turn)
```

코드 실행: [(trinket code) polygon_arc](https://trinket.io/python/1bcfca23780e)

**문제 3**

`polygon_arc()` 함수를 이용하여 호를 그리는 `arc(t, r, n, angle)` 함수를 구현하라.
네 매개 변수 각각에 입력되는 값의 의미는 다음과 같다.

- `t`: 거북이 객체
- `r`: 원의 반지름
- `n`: n-각형 지정
- `angle`: 호의 내각.

아래 그림은 반지름이 100인 원을 120도 만큼만, 즉 1/3 정도만 그린 결과를 보여준다.

```python
arc(t, 100, n=100, angle=120)
```

힌트: `n`각형을 그릴 때 한 변의 길이를 원 둘레를 이용하여 계산한다.

<p><div align="center" border="1px"><img src="https://raw.githubusercontent.com/codingalzi/42H/master/jupyter-book/images/turtle13-1.png" width="300"/></div></p>

답:

`circle()` 함수가 `polygon()` 함수를 이용하듯이
`arc()` 함수가 `polygon_arc()` 함수를 이용하면 되다. 
단, `angle` 매개 변수를 추가하기만 하면 된다.

```python
def arc(t, r, n=100, angle=360):
    length = (2 * math.pi * r) / n
    polygon_arc(t, length, n, angle)
```

**문제 4**

꽃잎 모양을 그리는 `leaf(t, r, n, angle)` 함수를 구현하라.
네 매개 변수 각각에 입력되는 값의 의미는 다음과 같다.

- `t`: 거북이 객체
- `r`: 원의 반지름
- `n`: n-각형 지정
- `angle`: 호의 내각.

예를 들어 `leaf(t, 100, n=300, angle=60)`를 실행하면 다음과 같이 그려져야 한다.

힌트: `arc()` 함수를 두 번 적용한다. 
단 한 번 적용한 다음에 반대 방향으로 회전을 먼저 해야 한다.

<p><div align="center" border="1px"><img src="https://raw.githubusercontent.com/codingalzi/42H/master/jupyter-book/images/turtle14-1.png" width="300"/></div></p>

답:

`arc(t, r, n, angle)`을 실행한 다음엔 `180-angle`만큼 반시계 방향으로 회전한 다음에
`arc(t, r, n, angle)`를 한 번 더 호출하면 출발했던 자리로 되돌아 온다.
마지막에 `180-angle` 만큼 한 번 더 반시계 방향으로 회전하면 출발할 때 가리켰던 방향을 다시 가리킨다.

설명한 내용을 이용하여 `leaf()` 함수를 다음과 같이 작성한다.
매개 변수 `n`과 `angle`은 각각 꽃잎의 길이와 너비를 조절한다.

```python
def leaf(t, r, n=300, angle=60):
    turn = 180-angle
    arc(t, r, n, angle)
    t.left(turn)
    arc(t, r, n, angle)
    t.left(turn)
```

**문제 5**

꽃을 그리는 `def flower(t, r, n, angle, leaves)` 함수를 구현하라.

- `t`: 거북이 객체
- `r`: 원의 반지름
- `n`: n-각형 지정
- `angle`: 호의 내각
- `leaves`: 꽃잎 개수

예를 들어 `flower(t, 100, 300, 60, 7)`를 실행하면 다음과 같이 그려져야 한다.

힌트: `leaf()` 함수 활용

<p><div align="center" border="1px"><img src="https://raw.githubusercontent.com/codingalzi/42H/master/jupyter-book/images/turtle15-1.png" width="300"/></div></p>

답:

`leaf()` 함수를 한 번 실행한 다음엔 `360/꽃잎수`만큼 반시계 방향으로 회전시키는 과정을 `꽃잎수`만큼 반복하면된다.
아래 코드에서 `leaves`가 꽃잎의 수를 인자로 받는 매개 변수다.

```python
def flower(t, r, n=300, angle=60, leaves=7):
    for _ in range(leaves):
        leaf(t, r, n, angle)
        t.left(360/leaves) # 회전 각도
```
