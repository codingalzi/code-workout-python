### 연습문제

**문제 1**

학생 정보를 표현하는 `Student` 클래스를 정의하라.
`Student` 클래스는 다음 두 개의 인스턴스 변수를 가져야 하며,
학생 점수는 부동소수점으로 저장되어야 한다.

| 인스턴스 변수 | 설명 |
|---|---|
| `self.name` | 학생 이름 |
| `self.score` | 학생 점수 |

단, 클래스 본문에 `__init__()` 메서드만 정의하여 아래 코드가 작동하도록 해야 한다.

```python
kgh = Student("김강현", 85)
print(kgh.name)
print(kgh.score)
```

힌트: `Student("김강현", 85)` 처럼 `Student` 클래스의 인스턴스를 생성할 때 
`"김강현"`과 `85` 두 개의 인자를 지정하려면
생성자 메서드가 `def __init__(self, name, score): ...` 헤더를 가져야 한다.
또한 함수 본문에서 `name`과 `score` 두 매개변수의 인자로 들어오는 값을 각각 서로 다른 
인스턴스 변수로 저장해야 한다.

답:

아래처럼 `Student` 클래스를 정의할 수 있다.
`score`는 `float()` 함수를 이용하여 부동소수점으로 저장한다.

````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = float(score)
````

아래 코드는 김강현의 이름과 점수를 저장한 객체를 생성한 후에 아래 두 속성을 출력한다.

- `student.name`: 문자열 `"김강현"`을 가리킴
- `student.score`: 부동소수점 `85.0`을 가리킴

````python
kgh = Student("김강현", 85)
print(kgh.name)
print(kgh.score)
````

따라서 두 학생의 `name`과 `score` 속성은 서로 독립적으로 저장된다.

**문제 3**

`Student` 클래스를 수정하여 점수를 항상 정수로 저장하도록 하라.
예를 들어 아래 코드 실행 후 `kgh.score`는 정수 `85`가 되어야 한다.


````python
kgh = Student("김강현", 85)
````


답:

`+` 연산자가 두 학생의 평균 점수를 반환하도록 하려면 `__add__()` 메서드를 정의한다.

````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = int(score)

    def __repr__(self):
        return f"Student: {self.name!r}"

    def __eq__(self, other):
        return self.score == other.score

    def __len__(self):
        return len(self.name)

    def __lt__(self, other):
        return self.score < other.score

    def __add__(self, other):
        return (self.score + other.score)/2

kgh = Student("김강현", 85)
whang = Student("황현", 91)
````

**문제 7**

`Student` 클래스에 `__getitem__()` 메서드를 추가하라.

인덱스 `0`은 학생 이름을, 인덱스 `1`은 학생 점수를 반환하도록 한다.

```python
kgh = Student("김강현", 85)

print(kgh[0])  # 김강현
print(kgh[1])  # 85
```

답:

대괄호 인덱싱을 지원하려면 `__getitem__()` 메서드를 정의한다.
인덱스가 0이면 이름을, 1이면 점수를 반환하도록 만들 수 있다.

````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = int(score)

    def __repr__(self):
        return f"Student: {self.name!r}"

    def __eq__(self, other):
        return self.score == other.score

    def __len__(self):
        return len(self.name)

    def __lt__(self, other):
        return self.score < other.score

    def __add__(self, other):
        return (self.score + other.score)/2

    def __bool__(self):
        return self.score >= 60

    def __getitem__(self, index):
        if index == 0:
            return self.name
        elif index == 1:
            return self.score
        else:
            raise IndexError("인덱스는 0 또는 1만 허용.")

kgh = Student("김강현", 85)
````

`kgh[0]`은 내부적으로 `kgh.__getitem__(0)`을 호출하고,
`kgh[1]`은 `kgh.__getitem__(1)`을 호출한다.

````python
print(kgh[0])
print(kgh[1])
````

## 인스턴스 메서드

### 연습문제

`Student` 클래스가 다음과 같다.

````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = int(score)

    def __repr__(self):
        return f"Student: {self.name!r}"

    def __eq__(self, other):
        return self.score == other.score

    def __len__(self):
        return len(self.name)

    def __lt__(self, other):
        return self.score < other.score

    def __add__(self, other):
        return (self.score + other.score) / 2

    def __bool__(self):
        return self.score >= 60

    def __getitem__(self, index):
        if index == 0:
            return self.name
        elif index == 1:
            return self.score
        else:
            raise IndexError("인덱스는 0 또는 1만 허용.")
````

이어지는 문제들에서 요구하는 기능을 수행하는 인스턴스 메서드를 `Student` 클래스에 하나씩 추가하라.

**문제 1**

`Student` 클래스에 `grade()` 인스턴스 메서드를 추가하라.

`grade()` 메서드는 점수에 따라 다음 문자열을 반환해야 한다.

| 점수 | 반환값 |
|---|---|
| 90점 이상 | `"A"` |
| 80점 이상 90점 미만 | `"B"` |
| 70점 이상 80점 미만 | `"C"` |
| 60점 이상 70점 미만 | `"D"` |
| 60점 미만 | `"F"` |

```python
whang = Student("황현", 91)
print(whang.grade())  # A
```

답:

`grade()` 메서드는 `self.score` 값을 조건문으로 검사하여 알맞은 등급 문자열을 반환하면 된다.

````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = int(score)

    def __repr__(self):
        return f"Student: {self.name!r}"

    def __eq__(self, other):
        return self.score == other.score

    def __len__(self):
        return len(self.name)

    def __lt__(self, other):
        return self.score < other.score

    def __add__(self, other):
        return (self.score + other.score) / 2

    def __bool__(self):
        return self.score >= 60

    def __getitem__(self, index):
        if index == 0:
            return self.name
        elif index == 1:
            return self.score
        else:
            raise IndexError("인덱스는 0 또는 1만 허용.")

    def grade(self):
        if self.score >= 90:
            return "A"
        elif self.score >= 80:
            return "B"
        elif self.score >= 70:
            return "C"
        elif self.score >= 60:
            return "D"
        else:
            return "F"

whang = Student("황현", 91)
````

`whang.score`는 `91`이므로 첫 번째 조건 `self.score >= 90`이 참이다.
따라서 `whang.grade()`는 `"A"`를 반환한다.

````python
print(whang.grade())
````

`update_score()` 메서드는 학생의 점수를 `new_score`로 변경해야 한다.
단, 새 점수는 정수로 저장되어야 한다.

답:

`update_score()` 메서드는 새 점수를 매개변수 `new_score`로 전달받아

`score_gap(other)` 메서드는 두 점수의 차이를 계산한 다음,

````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = int(score)

    def __repr__(self):
        return f"Student: {self.name!r}"

    def __eq__(self, other):
        return self.score == other.score

    def __len__(self):
        return len(self.name)

    def __lt__(self, other):
        return self.score < other.score

    def __add__(self, other):
        return (self.score + other.score) / 2

    def __bool__(self):
        return self.score >= 60

    def __getitem__(self, index):
        if index == 0:
            return self.name
        elif index == 1:
            return self.score
        else:
            raise IndexError("인덱스는 0 또는 1만 허용.")

    def grade(self):
        if self.score >= 90:
            return "A"
        elif self.score >= 80:
            return "B"
        elif self.score >= 70:
            return "C"
        elif self.score >= 60:
            return "D"
        else:
            return "F"

    def update_score(self, new_score):
        self.score = int(new_score)

    def is_better_than(self, other):
        return self.score > other.score

    def score_gap(self, other):
        return abs(self.score - other.score)

kgh = Student("김강현", 85)
whang = Student("황현", 91)
````

```python
subject_averages = []
```
