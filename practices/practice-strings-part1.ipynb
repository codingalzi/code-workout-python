## 빈 문자열

### 연습문제

**문제 1**

아래 코드가 `False`를 출력하는 이유를 설명하라.

````python
print('' == ' ')
````

답:

**문제 2**

행사장 입구에서 손님에게 순번 팔찌를 나눠 준다. 빈 문자열에서 시작하여 `for` 반복문으로 `손님1 손님2 손님3 ` 형태의 안내 문구를 만들어 출력하라.

답:

`for` 반복문과 문자열 이어붙이기 연산을 이용한다.

````python
message = ''
for i in range(3):
    message = message + '손님' + str(i + 1) + ' '

print(message)
````

## 날 문자열

### 연습문제

**문제 1**

아래 두 코드의 출력 차이를 설명하라.

````python
print('C:\new')
print(r'C:\new')
````

답:

일반 문자열에서는 `\n`이 줄바꿈으로 처리될 수 있지만, 날 문자열에서는 백슬래시와 `n`이 그대로 보존된다.

**문제 2**

다음 코드를 실행할 때 오류가 발생하는 이유를 설명하고 대안을 제시하라.

````python
print(r'C:\Users\student\')
````

답:

날 문자열도 마지막 문자를 백슬래시 하나로 끝낼 수 없다. 마지막 백슬래시가 닫는 따옴표에 영향을 주기 때문이다.
이런 경우엔 날 문자열대신 이스케이프 시퀀스를 활용할 수밖에 없다.

````python
print('C:\\Users\\student\\')
````

## f-문자열 서식 지정

### 연습문제

**문제 1**

문자열 `'hi'`를 6칸 폭에서 가운데 정렬하고, 빈 칸은 `-`로 채워 출력하라.

답:

````python
word = 'hi'
print(f'{word:-^6}')
````

**문제 2**

정수 `42`를 다섯 칸 폭으로 오른쪽 정렬하고, 빈 칸은 `0`으로 채워 출력하라.

답:

````python
num = 42
print(f'{num:0>5d}')
````

**문제 3**

부동소수점 `-84.99888`을 소수점 아래 셋째 자리까지 출력하라.

답:

소수점 이하 넷째 자리에서 반올림된다.

````python
num = -84.99888
print(f'{num:.3f}')
````

**문제 4**

작은 극장의 좌석표를 출력한다. 구역 이름 `A`는 4칸 가운데 정렬하고 빈 칸은 `*`로 채운다. 가격 `12000.5`는 10칸 폭에 소수점 아래 첫째 자리까지 출력하여 한 줄로 보여라.

답:

````python
zone = 'A'
price = 12000.5

print(f'구역:{zone:*^4} 가격:{price:10.1f}')
````
