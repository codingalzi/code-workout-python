**문제 2**

문자열에서 가장 긴 회문 부분 문자열을 찾고자 한다.

(1) 다음 문자열에서 가장 긴 회문 부분 문자열을 찾아 출력하는 코드를 작성하라.

```python
word = "abaxyzzyxf"
```

부분 문자열은 문자열에서 연속된 일부 문자열을 의미한다.
예를 들어 `"aba"`, `"xyzzyx"`는 모두 `word`의 부분 문자열이며, 둘 다 회문이다.

실행 결과는 다음과 같아야 한다.

```python
xyzzyx
```

힌트: 중첩 `for` 반복문을 이용하여 가능한 모든 부분 문자열을 슬라이싱으로 확인할 것.

답:

가능한 모든 부분 문자열을 슬라이싱으로 생성한 뒤,
각 부분 문자열이 자신의 역순과 같은지 비교한다.
그중 가장 긴 회문 문자열을 계속 갱신해 나가면 최종적으로 가장 긴 회문을 찾게 된다.

아래 코드에서 `start`와 `end`는 각각 부분 문자열의 시작과 끝 위치를 가리킨다.
`word[start:end]`는 `start`번 인덱스부터 `end - 1`번 인덱스까지의 부분 문자열을 생성한다.

생성된 부분 문자열 `sub`가 `sub[::-1]`과 같으면 회문이다.
이때 현재까지 찾은 회문 문자열 `longest`보다 길면 `longest`를 `sub`로 업데이트한다.

모든 부분 문자열을 확인한 뒤 `longest`를 출력하면 가장 긴 회문 부분 문자열이 확인된다.

````python
word = "abaxyzzyxf"

longest = ""

for start in range(len(word)):
    for end in range(start + 1, len(word) + 1):
        sub = word[start:end]

        if sub == sub[::-1] and len(sub) > len(longest):
            longest = sub

print(longest)
````

(2) 주어진 문자열에서 가장 긴 회문 부분 문자열을 찾는 함수 `finding_palindrome(word)`를 정의하라.

힌트: 이전 코드 활용

답:

이전 코드에서 `word`는 함수의 매개변수로, `longest` 함수의 반환값으로 지정하면 된다.

````python
def finding_palindrome(word):

    longest = ""

    for start in range(len(word)):
        for end in range(start + 1, len(word) + 1):
            sub = word[start:end]

            if sub == sub[::-1] and len(sub) > len(longest):
                longest = sub

    return longest
````

````python
finding_palindrome("abaxyzzyxf")
````

**문제 1**

변수와 함수의 이름을 지정할 때 **낙타**<font size = "2">camel</font> 표기법 또는
**팟홀**<font size="2">pothole</font> 표기법을 사용한다.

* 낙타 표기법: 소문자로 시작하고, 이어지는 단어의 시작은 대문자로 작성하는 표기법.
* 팟홀 표기법: 모두 소문자를 사용하고, 단어 사이에 밑줄기호(`_`)를 사용하는 표기법.

활용 예제는 다음과 같다.

- 낙타 표기법 예제: `userName`, `printMessage`, `countA` 등.
- 팟홀 표기법 예제: `user_name`, `print_message`, `count_a` 등.   


낙타 표기법으로 작성된 변수를 팟홀 표기법으로 변환하는 함수 `camel_to_snake(name)` 함수를 정의한 다음에 아래 변수를 변환하라.

```
CamelCase
```

힌트: `for` 반복문과 `isupper()`, `lower()` 메서드 활용

답:

낙타 표기법의 핵심은 변수에 사용된 두 단어를 대문자로 구분하는 데에 있다.
따라서 대문자 알파벳이 사용된 위치를 기준으로 밑줄을 추가하고 대문자는 소문자로 변환한다.
이 과정을 문자열에 대한 `for` 반복문으로 처리해서 새로운 문자열을 만들어 함수의
반환값으로 지정한다.

````python
def camel_to_snake(name):    
    result = ""                          # 변환된 문자열 누적 업데이트

    for char in name:
        if char.isupper():               # 대문자인 경우, 언더스코어와 소문자로 변환하여 누적
            result += "_" + char.lower()
        else:
            result += char

    if result.startswith("_"):           # 변환된 문자열이 언더스코어로 시작하는 경우 제거
        result = result[1:]

    return result
````

````python
camel_to_snake('CamelCase')
````

**문제 3**

문자열 `find()` 메서드에 대응하는 `find()` 함수를 직접 정의하고자 한다.

(1) `find(word1, word2)` 함수는 두 개의 매개변수를 사용하며 각 매개변수의 기능은 다음과 같아야 한다.

* 매개변수: 두 개의 문자열을 인자로 받는 `word1`과 `word2` 두 개 사용
* `word1`이 `word2`를 부분 문자열로 포함할 경우 부분 문자열이 시작하는 인덱스 반환.
    포함하지 않을 경우 -1을 반환.
    
단, 문자열 인덱싱, 슬라이싱만 사용하고 다른 문자열 메서드는 전혀 사용하지 않아야 한다.

답:

예를 들어 `'banana'` 문자열은 `'ana'` 문자열을 부분 문자열로 포함한다.
`'ana'` 부분 문자열이 `'banana'` 문자열의 어느 인덱스에서 시작하는지를
확인하려면 모든 인덱스에 대해 그곳에서부터 `'ana'` 문자열의 시작하는지 여부를
슬라이싱으로 대조하면 된다.

예를 들어 1번 인덱스부터 `'ana'` 문자열이 시작하는지 여부는 다음과 같이 확인한다.
아래 코드에서 3은 `'ana'` 문자열의 길이를 나타낸다.

```python
'banana'[1:1+3] == 'ana'
```

아래 `find()` 함수는 `word1` 문자열이 `word2` 문자열을 부분 문자열로 포함되는지 여부를
판단하기 위해 `word1`의 모든 인덱스에 대해 `word2` 부분 문자열과 일치 여부를
슬라이싱을 이용하여 확인하는 순간에 해당 인덱스를 반환한다.
또한 부분 문자열을 확인하지 못하면 -1을 반환한다.

````python
def find(word1, word2):
    for idx in range(len(word1)):
        if word2 == word1[idx:idx+len(word2)] :
            return idx
    return -1
````

````python
print(find('banana', 'ana'))
print(find('heyheyhey', 'ey'))
print(find('waikiki', 'ki'))
print(find('ulala', 'la'))
print(find('chukachuka', 'uKa'))
````

(2) 이번에는 `find(word1, word2, position=0)` 함수가 세 개의 매개변수를 사용하며 각 매개변수의 기능은 다음과 같도록 정의하라.

* 매개 변수: 두 개의 문자열을 인자로 받는 `word1`과 `word2` 두 개와 키워드 인자를 받는 `position`.
* `word1`이 `word2`를 부분 문자열로 포함할 경우 부분 문자열이 시작하는 인덱스 반환.
    포함하지 않을 경우 -1을 반환.
* `position` 매개변수: 탐색을 시작하는 인덱스 지정. 기본 키워드 인자로 0 사용.

단, 문자열 인덱싱, 슬라이싱만 사용하고 다른 문자열 메서드는 전혀 사용하지 않아야 한다.

힌트: `range()` 함수를 활용한 탐색 구간 지정

**답**

`find()` 함수의 헤더를 다음과 같이 지정한다.

```python
def find(word1, word2, position=0)
```

`position=0`은 부분 문자열 탐색을 시작하는 인덱스를 가리키며 기본값으로 0을 사용한다.
즉, `position` 값을 지정하지 않거나 변경하지 않는다면 `word1`이 가리키는
문자열의 전 구간을 대상으로 부분 문자열 탐색 여부를 확인한다.

반면에 `position` 값을 0이 아닌 다른 값으로 지정하면
해당 인덱스부터 탐색을 시작한다.
이를 위해 `range()` 함수가 지정하는 구간을 `position`으로 시작하도록 한다.

```python
range(position, len(word1))
```

`find()` 함수의 다른 부분은 동일하며, 정리하면 다음과 같다.

````python
def find(word1, word2, position=0):
    for idx in range(position, len(word1)):
        if word2 == word1[idx:idx+len(word2)] :
            return idx
    return -1
````

- `position=0`을 사용하는 경우

````python
print(find('banana', 'ana'))
print(find('heyheyhey', 'ey'))
print(find('waikiki', 'ki'))
print(find('ulala', 'la'))
print(find('chukachuka', 'uKa'))
````

- `position`을 다르게 지정하는 경우

````python
print(find('banana', 'ana', 2))
print(find('heyheyhey', 'ey', 3))
print(find('waikiki', 'ki', 4))
print(find('ulala', 'la', 2))
print(find('chukachuka', 'uKa'))
````
