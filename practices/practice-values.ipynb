### 연습문제

**문제 1**

체질량지수(BMI, Body Mass Index)는 키와 체중을 이용해 비만도를 간단히 평가하는 지표로, 
체중(kg)을 키(m)의 제곱으로 나눈 값이다.
즉, 다음 식으로 체질량지수를 계산한다.

```
BMI = 체중 ÷ (키²)
```

몸무게가 80kg이고 키가 1.8m인 사람의 BMI를 계산하여 다음과 같이 출력하는 코드를 작성하라.

```
키 1.8 m, 몸무게 80 kg인 사람의 BMI: 24.691358024691358
```

힌트: `weight`, `height`, `bmi` 세 개의 변수와 거듭제곱 연산자 `**`를 활용한다.

답:

````python
weight = 80
height = 1.8
bmi = weight / (height ** 2)
print("키", height, "m, 몸무게", weight, "kg인 사람의 BMI:", bmi)
````

**문제 3**

키(신장)가 120cm 이상이어야 탑승이 가능한 놀이기구가 있다. 
현재 키가 112cm인 어린이가 매년 2.5cm씩 자란다고 가정할 때, 
3년 뒤에는 이 놀이기구에 탑승할 수 있는지 여부를 가리키는 변수 `can_ride`를 활용하여 판단하는 코드를 작성하라.

답:

````python
limit = 120           # 놀이기구 탑승 최소 신장(cm)
current_height = 112  # 현재 신장(cm)
growth_rate = 2.5     # 연간 신장 성장률(cm)
years = 3             # 성장 기간(년)

future_height = current_height + (growth_rate * years) # 미래 신장(cm)
can_ride = future_height >= limit                      # 탑승 가능 여부

print(can_ride)
````

## 연산자 우선순위

### 연습문제

**문제 1**

다음 수식의 계산 결과를 예측하고, 연산 순서를 괄호 `()`를 사용하여 그 이유를 설명하라

````python
7 + 3 * 2 ** 2 == 7 + (3 * (2 ** 2))
````

답:

결과는 `True`이다.
이유는 동치 비교 연산자의 좌우 모두 19로 계산되기 때문이다. 

- 연산 순서는 거듭제곱(`**`) $\rightarrow$ 곱셈(`*`) $\rightarrow$ 덧셈(`+`) 순이다.
- 따라서 `7 + 3 * 2 ** 2`를 괄호로 표현하면 `7 + (3 * (2 ** 2))`이 된다.

````python
7 + 3 * 2 ** 2 == 7 + (3 * (2 ** 2))
````

**문제 4**

리스트 `data = [10, 20, 30]`이 주어졌을 때, 첫 번째 항목과 두 번째 항목의 평균을 구하는 코드를 작성하라. 
단, 괄호를 적절히 사용하여 우선순위를 제어해야 한다.

답:

````python
data = [10, 20, 30]
average = (data[0] + data[1]) / 2
print(average)
````

덧셈을 먼저 수행하기 위해 분자를 괄호로 감싸야 한다.
만약 괄호를 쓰지 않고 `data[0] + data[1] / 2`라고 쓰면 나눗셈이 먼저 수행되어 `10 + 10.0` 즉, 20.0이 되어 오답이 된다.

````python
data = [10, 20, 30]
average = data[0] + data[1] / 2
print(average)
````

````python
(25000 + 19000) / 4
````
