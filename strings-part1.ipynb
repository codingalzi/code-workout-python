## 빈 문자열

어떤 문자도 포함하지 않는 문자열을 **빈 문자열**<font size="2">empty string</font>이라 한다.
빈 문자열은 다음 세 가지 방식으로 만들 수 있다.

````python
empty_str1 = ''
empty_str2 = ""
empty_str3 = str()

print(empty_str1 == empty_str2 == empty_str3)
````

빈 문자열을 출력하면 화면에는 아무것도 보이지 않는다.
하지만 값이 없는 것이 아니라, 길이가 0인 문자열 값이다.

````python
print(empty_str1)
print(len(empty_str1))
````

반면에 `' '`는 공백 문자 하나를 포함한 문자열이다.
눈에 잘 보이지 않지만 길이는 1이다.

````python
blank = ' '
print(len(blank))
````

:::{note} 한 글자 문자열

C, 자바 같은 언어는 문자 하나를 저장하는 `char` 자료형을 별도로 제공한다.
파이썬에는 별도의 문자 자료형이 없으며, 문자 하나도 길이가 1인 문자열로 다룬다.
:::

## 화이트 스페이스

**화이트 스페이스**<font size="2">whitespace</font>는 화면에 직접 보이지는 않지만
문자열 안에서 공백, 줄바꿈, 간격 등을 나타내는 문자다.
다음 세 가지 화이트 스페이스를 가장 많이 사용한다.

| 표기 | 의미 |
| :---: | :--- |
| `' '` | 공백 문자 |
| `'\t'` | 탭 |
| `'\n'` | 줄바꿈 |

````python
print('A B')
print('A\tB')
print('A\nB')
````

화이트 스페이스도 문자열에 포함된 하나의 문자로 계산된다.

````python
print(len('A B'))
print(len('A\tB'))
print(len('A\nB'))
````

## 이스케이프 시퀀스

문자열 안에서 따옴표, 백슬래시, 줄바꿈처럼 특별한 의미를 갖는 문자를 표현하려면
백슬래시(`\`)로 시작하는 **이스케이프 시퀀스**<font size="2">escape sequence</font>를 사용한다.

| 표기 | 문자열에 들어가는 문자 |
| :---: | :--- |
| `\'` | 작은따옴표 |
| `\"` | 큰따옴표 |
| `\\` | 백슬래시 자체 |
| `\n` | 줄바꿈 |
| `\t` | 탭 |

문자열을 감싼 따옴표와 같은 따옴표를 문자열 안에 넣을 때는 이스케이프 시퀀스가 필요하다.

````python
sentence1 = 'I\'m learning Python.'
sentence2 = "그가 \"좋아요\"라고 말했다."

print(sentence1)
print(sentence2)
````

백슬래시를 생략하면 문제가 발생할 수 있다.
문자열을 감싼 따옴표와 동일한 따옴표를 문자열 안에 사용하면 `SyntaxError` 오류가 발생한다.

````python
sentence1 = 'I'm learning Python.'

print(sentence1)
````

감싸는 따옴표를 문자열 내부에 사용된 따옴표와 다르면 백슬래시 없어도 오류가 발생하지 않는다.

````python
sentence3 = "I'm learning Python."
print(sentence3)
````

또한 백슬래시 자체를 문자열에 넣으려면 `\\`처럼 두 번 적는다.

````python
path = 'C:\\Users\\student\\notes.txt'

print(path)
````

:::{admonition} 백슬래시와 원화 기호
:class: note

한국어 윈도우 환경에서는 백슬래시(`\`)가 원화 기호처럼 보일 수 있다.
화면에 보이는 모양이 달라도 파이썬 코드에서의 기능은 동일하다.
:::

## 날 문자열

문자열 앞에 `r` 또는 `R`을 붙이면 **날 문자열**<font size="2">raw string</font>이 된다.
날 문자열에서는 `\n`, `\t` 같은 이스케이프 시퀀스가 특별한 기능으로 해석되지 않고
대부분 그대로 보존된다.

````python
normal = 'Python\nis\teasy'
raw = r'Python\nis\teasy'

print(normal)
print(raw)
````

날 문자열은 특히 파일 경로나 정규표현식처럼 백슬래시가 자주 등장하는 문자열에서 유용하게 활용된다.

````python
print('C:\\some\\name')
print(r'C:\some\name')
````

## f-문자열 서식 지정

f-문자열은 값을 문자열 안에 넣는 데 그치지 않고, 출력 폭과 정렬 방식, 소수점 자릿수도 지정할 수 있다.
서식 지정은 중괄호 안에서 콜론(`:`) 뒤에 작성한다.

```python
f'{표현식:서식}'
```

### 정렬과 폭

문자열을 일정한 칸 안에 배치하려면 폭을 지정한다.
`<`, `>`, `^`는 각각 왼쪽, 오른쪽, 가운데 정렬을 의미한다.

````python
s1 = 'hi'
s2 = 'hello'

print(f'[{s1:<10}]')
print(f'[{s2:<10}]')
print(f'[{s1:>10}]')
print(f'[{s2:>10}]')
print(f'[{s1:^10}]')
print(f'[{s2:^10}]')
````

남는 칸을 공백이 아닌 다른 문자로 채울 수도 있다.
채움 문자는 정렬 기호 앞에 적는다.

````python
print(f'{s1:+<10}')
print(f'{s2:0>10}')
print(f'{s1:-^10}')
````

### 소수점 자릿수

부동소수점을 출력할 때 `{값:.nf}` 형식을 사용하면 소수점 아래 `n`자리까지 반올림해서 보여준다.

````python
num1 = 17.153742
num2 = 2.778

print(f'{num1:.1f}')
print(f'{num1:.2f}')
print(f'{num1:.3f}')
print(f'{num2:.2f}')
````

폭 지정과 소수점 자릿수 지정은 함께 사용할 수 있다.

````python
print(f'[{num1:10.2f}]')
print(f'[{num2:10.2f}]')
print(f'[{num1:^10.2f}]')
print(f'[{num2:^10.2f}]')
````

### 연습문제

**문제 1**

아래 두 정수를 각각 네 칸을 차지하도록 오른쪽 정렬하고, 빈 칸은 `0`으로 채워라.

````python
num1 = 7
num2 = 42
````

답:

`0>4d`는 정수를 네 칸 폭으로 오른쪽 정렬하고, 남는 왼쪽 칸을 `0`으로 채운다는 뜻이다.

````python
print(f"{num1:0>4d}")
print(f"{num2:0>4d}")
````

**문제 2**

아래 두 점수를 소수점 아래 첫째 자리까지 출력하되, 전체 폭은 8칸으로 맞춰라.

````python
score1 = 93.456
score2 = 7.8
````

답:

`8.1f`는 전체 폭을 8칸으로 잡고, 소수점 아래 첫째 자리까지 반올림해 출력한다.

````python
print(f"{score1:8.1f}")
print(f"{score2:8.1f}")
````

**문제 4**

아래 부동소수점을 소수점 이하 셋째 자리까지만 출력되도록 f-문자열을 활용하라.

````python
num = 23.36778
num1 = -84.99888
````

답:

아래 코드는 소수점 이하 넷째 자리에서 반올림한 값을 출력한다.

````python
print(f"{num:.3f}")
print(f"{num1:.3f}")
````

**문제 5**

아래 부동소수점의 소수점을 모두 버리로 정수 부문만 출력되도록 f-문자열을 활용하라.
또한 정수는 총 10칸을 차지하도록 한다.
단, `int()` 함수는 사용하지 않는다.

````python
num2 = 23.36778
num3 = 8223.62874636778
````

답:

아래 코드는 소수점 이하 첫째 자리에서 반올림한 값을 출력하며,
총 10칸을 차지하도록 하며 오른쪽으로 정렬시킨다.

````python
print(f"{num2:10.0f}")
print(f"{num3:10.0f}")
````

**문제 6**

아래 두 개의 부동소수점을 총 15 칸을 차지하도록 f-문자열을 활용하라.
또한 소수점 이하 다섯째 자리까지만 출력되록 한다.
단 모든 정수는 오른쪽으로 정렬되어야 하며 앞쪽 빈칸은 `+` 기호로 채운다.

````python
num4 = 887223.8348161
num5 = 6833.127230009267
````

답:

아래 코드는 소수점 이하 여섯째 자리에서 반올림한 값을 출력하며,
총 15칸을 차지하도록 하며 오른쪽으로 정렬시킨다.
또한 왼쪽의 빈 칸은 `+` 기호로 채운다.

````python
print(f"{num3:+>15.5f}")
print(f"{num4:+>15.5f}")
````
